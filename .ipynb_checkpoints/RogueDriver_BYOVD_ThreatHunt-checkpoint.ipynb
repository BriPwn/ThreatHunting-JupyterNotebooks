## 📋 Overview

This notebook provides threat hunting and detection strategies for **Rogue Driver Installation** via the **Bring Your Own Vulnerable Driver (BYOVD)** technique. BYOVD is one of the most prevalent defense-evasion primitives in modern intrusions — used by ransomware gangs (RansomHub, LockBit, BlackCat), nation-state actors (Scattered Spider, VENOMOUS BEAR/Turla), and commodity toolkits alike.

### How BYOVD Works

Attackers bring a **legitimately signed but vulnerable driver** onto the target, load it as a kernel service, then exploit its vulnerability to gain kernel read/write primitives:

| Goal | Example Action |
|------|---------------|
| Kill EDR/AV | Terminate protected sensor processes |
| Disable DSE | Turn off Driver Signature Enforcement, load unsigned rootkit |
| Remove EDR callbacks | Unregister kernel notification routines |
| Dump LSASS | Read process memory as kernel |

### Real-World Examples

| Driver | CVE | Abused By |
|--------|-----|-----------|
| `iqvw64.sys` (Intel NIC diag) | CVE-2015-2291 | Scattered Spider |
| `zamguard64.sys` / `zam64.sys` | — | BlackCat / Terminator tool |
| `aswArPot.sys` (Avast Anti-Rootkit) | — | Ransomware campaigns (Nov 2024) |
| `mhyprot2.sys` (Genshin Impact anti-cheat) | — | Multiple ransomware groups |
| `truesight.sys` (TrueSight) | — | EDRKillShifter / RansomHub |
| `PROCEXP152.sys` (Process Explorer) | — | Malvertising campaign |

## 🔴 Attack Phases

### Phase 1 — Drop the Vulnerable Driver
```
C:\Users\Public\iqvw64.sys
C:\Windows\Temp\zamguard64.sys
C:\ProgramData\aswArPot.sys
```

### Phase 2 — Load as a Kernel Service
```cmd
sc create VulnDrv binPath= C:\Users\Public\iqvw64.sys type= kernel start= demand
sc start VulnDrv
```
Tools like **KDMapper** bypass the SCM entirely — leaving no Event 7045.

### Phase 3 — Exploit Kernel R/W Primitive
The attacker sends `DeviceIoControl` calls to the loaded driver, obtaining arbitrary kernel read/write.

### Phase 4 — Payload (EDR Kill / DSE Disable / Rootkit)
```
Patch nt!g_CiEnabled = 0              // Disable Code Integrity
ZwTerminateProcess(EDR_PID)           // Kill the EDR sensor
NtLoadDriver(malicious_unsigned.sys)  // Load unsigned rootkit
```

### Key Tooling

| Tool | Purpose |
|------|---------|
| **KDMapper** | Maps drivers into kernel memory, bypasses SCM |
| **Terminator** (spyboy) | BYOVD wrapper targeting 23+ AV/EDR products |
| **EDRKillShifter** (RansomHub) | Automated BYOVD loader via ransomware affiliate program |
| **AuKill** (LockBit) | Custom BYOVD tool based on Backstab |
| **LIGHTSHOW** (UNC2970) | Nation-state BYOVD using ENE Technology driver |

````python
# ══════════════════════════════════════════════════════════════════════════════
# SIEM CONNECTION CONFIGURATION  —  update before running
# ══════════════════════════════════════════════════════════════════════════════

# ── Cortex XSIAM ──────────────────────────────────────────────────────────────
# API key: Settings → Configurations → Integrations → API Keys
XSIAM_FQDN       = "your-tenant.xdr.us.paloaltonetworks.com"
XSIAM_API_KEY    = "YOUR_XSIAM_API_KEY"
XSIAM_API_KEY_ID = "1"          # Numeric ID from XSIAM API key settings
XSIAM_LOOKBACK   = 7 * 24 * 60  # Lookback in minutes (default: 7 days)

# ── CrowdStrike NG-SIEM (LogScale / Falcon Long Term Repository) ───────────────
# API client scope required: Falcon LogScale (NG-SIEM) → Read
CS_CLIENT_ID     = "YOUR_CROWDSTRIKE_CLIENT_ID"
CS_CLIENT_SECRET = "YOUR_CROWDSTRIKE_CLIENT_SECRET"
CS_BASE_URL      = "https://api.crowdstrike.com"  # US-1; US-2: api.us-2.crowdstrike.com
CS_REPOSITORY    = "base_1"      # Your NG-SIEM repository name
CS_LOOKBACK      = "7d"          # Relative time string: "24h", "7d", "30d"

# ── Elastic / OpenSearch ──────────────────────────────────────────────────────
# API key: Stack Management → API Keys
ES_HOST          = "https://your-elastic-host:9200"
ES_API_KEY       = "YOUR_ELASTIC_API_KEY"    # base64-encoded  id:key  pair
ES_INDEX_PATTERN = "logs-*,winlogbeat-*,.ds-logs-*"
ES_LOOKBACK_DAYS = 7
ES_VERIFY_CERTS  = False  # Set True in production with a valid cert

# ── Microsoft Sentinel (Azure Monitor / Log Analytics) ────────────────────────
# App Registration in Entra ID with Log Analytics Reader on the workspace
SENTINEL_TENANT_ID     = "YOUR_AAD_TENANT_ID"
SENTINEL_CLIENT_ID     = "YOUR_APP_CLIENT_ID"
SENTINEL_CLIENT_SECRET = "YOUR_APP_CLIENT_SECRET"
SENTINEL_WORKSPACE_ID  = "YOUR_LOG_ANALYTICS_WORKSPACE_ID"
SENTINEL_LOOKBACK_DAYS = 7

print("\u2705  Configuration loaded.")
````

### Step 3 — Initialize Clients & Helper Functions

Run this cell once. All detection cells call these shared helpers.

````python
import hashlib, random, string, time, json, warnings, requests
from datetime import timedelta
from elasticsearch import Elasticsearch
from azure.identity import ClientSecretCredential
from azure.monitor.query import LogsQueryClient, LogsQueryStatus
from tabulate import tabulate

warnings.filterwarnings("ignore")

# ══════════════════════════════════════════════════════════════════════════════
# XSIAM helpers
# ══════════════════════════════════════════════════════════════════════════════
def _xsiam_headers():
    nonce = "".join(random.choices(string.ascii_letters + string.digits, k=64))
    ts    = str(int(time.time() * 1000))
    h     = hashlib.sha256((XSIAM_API_KEY + nonce + ts).encode()).hexdigest()
    return {
        "x-xdr-auth-id":   str(XSIAM_API_KEY_ID),
        "x-xdr-nonce":     nonce,
        "x-xdr-timestamp": ts,
        "Authorization":   f"Advanced {h}",
        "Content-Type":    "application/json",
    }

def run_xsiam(query, label='XSIAM'):
    start_url = f'https://{XSIAM_FQDN}/public_api/v1/xql/start_xql_query'
    payload = {
        "request_data": {
            "query": query,
            "timeframe": {"relativeTime": XSIAM_LOOKBACK * 60 * 1000},
        }
    }
    r = requests.post(start_url, json=payload, headers=_xsiam_headers(), timeout=30)
    r.raise_for_status()
    eid = r.json()["reply"]["execution_id"]

    poll_url = f'https://{XSIAM_FQDN}/public_api/v1/xql/get_xql_query_results'
    for _ in range(30):
        time.sleep(3)
        r = requests.post(
            poll_url,
            json={"request_data": {"execution_id": eid, "format_as_array": True}},
            headers=_xsiam_headers(),
            timeout=30,
        )
        r.raise_for_status()
        reply = r.json()["reply"]
        if reply["status"] == "SUCCESS":
            rows = reply.get("results", {}).get("data", [])
            _display(label, rows)
            return rows
        elif reply["status"] in ("FAILED", "CANCELLED"):
            raise RuntimeError(f'XQL query failed: {reply}')
    raise TimeoutError('XSIAM query timed out after 90 seconds')


# ══════════════════════════════════════════════════════════════════════════════
# CrowdStrike NG-SIEM helpers
# ══════════════════════════════════════════════════════════════════════════════
_cs_cache = {"token": None, "expires": 0}

def _get_cs_token():
    if time.time() < _cs_cache['expires'] - 60:
        return _cs_cache['token']
    r = requests.post(
        f'{CS_BASE_URL}/oauth2/token',
        data={"client_id": CS_CLIENT_ID, "client_secret": CS_CLIENT_SECRET},
        headers={"Content-Type": "application/x-www-form-urlencoded"},
        timeout=30,
    )
    r.raise_for_status()
    d = r.json()
    _cs_cache.update({
        "token":   d["access_token"],
        "expires": time.time() + d.get("expires_in", 1800),
    })
    return _cs_cache['token']

def run_crowdstrike(query, label='CrowdStrike NG-SIEM'):
    token = _get_cs_token()
    url   = f'{CS_BASE_URL}/humio-api/v1/repositories/{CS_REPOSITORY}/query'
    headers = {
        "Authorization": f"Bearer {token}",
        "Content-Type":  "application/json",
        "Accept":        "application/json",
    }
    payload = {"queryString": query, "start": CS_LOOKBACK, "end": "now", "isLive": False}
    r = requests.post(url, json=payload, headers=headers, timeout=60)
    r.raise_for_status()
    data = r.json()
    rows = data if isinstance(data, list) else data.get("results", data.get("events", []))
    _display(label, rows)
    return rows


# ══════════════════════════════════════════════════════════════════════════════
# Elastic helpers
# ══════════════════════════════════════════════════════════════════════════════
_es_client = None

def _get_es():
    global _es_client
    if _es_client is None:
        _es_client = Elasticsearch(
            ES_HOST, api_key=ES_API_KEY,
            verify_certs=ES_VERIFY_CERTS, ssl_show_warn=False,
        )
    return _es_client

def run_elastic(lucene_query, label='Elastic', size=50):
    client = _get_es()
    body = {
        "query": {
            "bool": {
                "must": [
                    {"query_string": {"query": lucene_query}},
                    {"range": {"@timestamp": {
                        "gte": f"now-{ES_LOOKBACK_DAYS}d", "lte": "now"}}},
                ]
            }
        },
        "size": size,
        "sort": [{"@timestamp": {"order": "desc"}}],
    }
    resp  = client.search(index=ES_INDEX_PATTERN, body=body)
    total = resp["hits"]["total"]["value"]
    rows  = [h["_source"] for h in resp["hits"]["hits"]]
    print(f'\n[{label}] Total index matches: {total}  (showing up to {size})')
    _display(label, rows)
    return rows


# ══════════════════════════════════════════════════════════════════════════════
# Microsoft Sentinel helpers
# ══════════════════════════════════════════════════════════════════════════════
_sentinel_client = None

def _get_sentinel():
    global _sentinel_client
    if _sentinel_client is None:
        cred = ClientSecretCredential(
            tenant_id=SENTINEL_TENANT_ID,
            client_id=SENTINEL_CLIENT_ID,
            client_secret=SENTINEL_CLIENT_SECRET,
        )
        _sentinel_client = LogsQueryClient(cred)
    return _sentinel_client

def run_sentinel(kql_query, label='Microsoft Sentinel'):
    client = _get_sentinel()
    result = client.query_workspace(
        workspace_id=SENTINEL_WORKSPACE_ID,
        query=kql_query,
        timespan=timedelta(days=SENTINEL_LOOKBACK_DAYS),
    )
    if result.status == LogsQueryStatus.SUCCESS:
        tbl  = result.tables[0]
        rows = [dict(zip(tbl.columns, row)) for row in tbl.rows]
        _display(label, rows)
        return rows
    else:
        print(f'\u274c  [{label}] Query error: {result.partial_error}')
        return []


# ══════════════════════════════════════════════════════════════════════════════
# Shared display helper
# ══════════════════════════════════════════════════════════════════════════════
def _display(label, rows, max_rows=25, max_col=90):
    if not rows:
        print(f'\n\u2705  [{label}]  No results — no hits in the lookback window.')
        return
    print(f'\n\U0001f6a8  [{label}]  {len(rows)} result(s) found!\n')
    if rows and isinstance(rows[0], dict):
        display = [{k: str(v)[:max_col] for k, v in row.items()} for row in rows[:max_rows]]
        print(tabulate(display, headers='keys', tablefmt='grid'))
        if len(rows) > max_rows:
            print(f'  ... ({len(rows) - max_rows} additional rows truncated)')
    else:
        for r in rows[:max_rows]:
            print(r)

print("\u2705  Helper functions initialized. Proceed to connection tests or detections.")
````

````python
# ── Cortex XSIAM connection test ──────────────────────────────────────────────
print("=" * 60)
print("Testing Cortex XSIAM...")
print("=" * 60)
try:
    test_q = 'dataset = xdr_data | limit 1 | fields agent_hostname'
    url = f'https://{XSIAM_FQDN}/public_api/v1/xql/start_xql_query'
    payload = {"request_data": {"query": test_q, "timeframe": {"relativeTime": 3600000}}}
    r = requests.post(url, json=payload, headers=_xsiam_headers(), timeout=15)
    r.raise_for_status()
    eid = r.json()['reply']['execution_id']
    print(f'\u2705  XSIAM connected  |  tenant: {XSIAM_FQDN}')
    print(f'    execution_id: {eid}')
except Exception as e:
    print(f'\u274c  XSIAM failed: {e}')

# ── CrowdStrike NG-SIEM connection test ───────────────────────────────────────
print()
print("=" * 60)
print("Testing CrowdStrike NG-SIEM...")
print("=" * 60)
try:
    tok = _get_cs_token()
    ttl = int(_cs_cache['expires'] - time.time())
    print(f'\u2705  CrowdStrike OAuth2 token obtained  |  TTL: ~{ttl}s')
    url = f'{CS_BASE_URL}/humio-api/v1/repositories/{CS_REPOSITORY}/query'
    r = requests.post(
        url,
        json={"queryString": "* | limit 1", "start": "1h", "end": "now", "isLive": False},
        headers={"Authorization": f"Bearer {tok}", "Content-Type": "application/json"},
        timeout=15,
    )
    r.raise_for_status()
    print(f'\u2705  LogScale repository "{CS_REPOSITORY}" accessible')
except Exception as e:
    print(f'\u274c  CrowdStrike NG-SIEM failed: {e}')

# ── Elastic connection test ────────────────────────────────────────────────────
print()
print("=" * 60)
print("Testing Elastic...")
print("=" * 60)
try:
    es = _get_es()
    if es.ping():
        info = es.info()
        print(f'\u2705  Elastic connected  |  cluster: {info["cluster_name"]}')
        print(f'    version: {info["version"]["number"]}')
    else:
        print('\u274c  Elasticsearch ping failed')
except Exception as e:
    print(f'\u274c  Elastic failed: {e}')

# ── Microsoft Sentinel connection test ────────────────────────────────────────
print()
print("=" * 60)
print("Testing Microsoft Sentinel...")
print("=" * 60)
try:
    sc = _get_sentinel()
    result = sc.query_workspace(
        workspace_id=SENTINEL_WORKSPACE_ID,
        query='union isfuzzy=true SecurityEvent | limit 1 | project TimeGenerated',
        timespan=timedelta(days=1),
    )
    if result.status == LogsQueryStatus.SUCCESS:
        print(f'\u2705  Sentinel connected  |  workspace: {SENTINEL_WORKSPACE_ID[:8]}...')
    else:
        print(f'\u26a0\ufe0f  Sentinel reachable but query returned: {result.status}')
except Exception as e:
    print(f'\u274c  Sentinel failed: {e}')
````

## 🔍 Detection #1: Kernel-Mode Service Installed from Suspicious Path

**Windows Event 7045** fires when a new service (including kernel drivers) is installed via the SCM. Hunting for `type=kernel` installs outside of standard deployment paths is high-fidelity for BYOVD.

> **Note**: KDMapper bypasses the SCM — it will **not** produce Event 7045. See Detection #3.

````python
# ── XSIAM — Detection #1 ───────────────────────────────────────────────────────
QUERY = '''
dataset = xdr_data
| filter event_type = WINDOWS and event_sub_type = WINDOWS_SYSTEM
| filter action_evtlog_event_id = 7045
| filter action_evtlog_data_fields["ServiceType"] ~= "(?i)kernel"
| filter action_evtlog_data_fields["ImagePath"] ~= "(?i)(\\\\Users\\\\|\\\\Temp\\\\|\\\\ProgramData\\\\|\\\\AppData\\\\)"
| fields _time, agent_hostname,
         action_evtlog_data_fields["ServiceName"] as ServiceName,
         action_evtlog_data_fields["ImagePath"]   as DriverPath,
         action_evtlog_data_fields["ServiceType"] as ServiceType,
         action_evtlog_data_fields["AccountName"] as InstalledBy
| sort desc _time
'''
run_xsiam(QUERY, label='Det-1 Kernel Service Install [XSIAM]')
````

````python
# ── CrowdStrike NG-SIEM — Detection #1 ─────────────────────────────────────────
QUERY = '''
#event_simpleName=DriverLoad OR #event_simpleName=ServiceInstalled
| ServiceType = /kernel/i
| ImageFileName = /(Users|Temp|ProgramData|AppData)/i
| table([timestamp, ComputerName, UserName, ServiceName, ImageFileName, SHA256HashData])
| sort(timestamp, order=desc)
'''
run_crowdstrike(QUERY, label='Det-1 Kernel Service Install [CrowdStrike]')
````

````python
# ── Elastic — Detection #1 ──────────────────────────────────────────────────────
QUERY = (
    'event.code:"7045" AND '
    'winlog.event_data.ServiceType:*kernel* AND ('
    '  winlog.event_data.ImagePath:*\\\\Users\\\\* OR '
    '  winlog.event_data.ImagePath:*\\\\Temp\\\\* OR '
    '  winlog.event_data.ImagePath:*\\\\ProgramData\\\\* OR '
    '  winlog.event_data.ImagePath:*\\\\AppData\\\\*'
    ')'
)
run_elastic(QUERY, label='Det-1 Kernel Service Install [Elastic]')
````

````python
# ── Microsoft Sentinel — Detection #1 ──────────────────────────────────────────
QUERY = '''
Event
| where EventID == 7045
| extend ServiceType = extract(@"ServiceType\\s*:\\s*([\\w ]+)",  1, EventData)
| extend ImagePath   = extract(@"ImagePath\\s*:\\s*([^\\r\\n]+)",  1, EventData)
| extend ServiceName = extract(@"ServiceName\\s*:\\s*([^\\r\\n]+)", 1, EventData)
| where ServiceType has "kernel"
| where ImagePath has_any (@"\\Users\\", @"\\Temp\\", @"\\ProgramData\\", @"\\AppData\\")
| project TimeGenerated, Computer, ServiceName, ServiceType, ImagePath, Account
| order by TimeGenerated desc
'''
run_sentinel(QUERY, label='Det-1 Kernel Service Install [Sentinel]')
````

## 🔍 Detection #2: Known Vulnerable Driver Hash Match

Matching Sysmon Event 6 (Driver Loaded) against the [LOLDrivers](https://www.loldrivers.io/) known-vulnerable hash list is the **highest-confidence** signal available. The cell below fetches the live feed automatically.

> **Tip**: Re-run the hash refresh cell periodically to stay current with new LOLDriver disclosures.

````python
# ── Fetch live LOLDrivers hash list ────────────────────────────────────────────
import urllib.request, json as _json

LOLDRIVERS_URL = "https://www.loldrivers.io/api/drivers.json"

try:
    print(f'Fetching LOLDrivers feed from {LOLDRIVERS_URL} ...')
    with urllib.request.urlopen(LOLDRIVERS_URL, timeout=30) as resp:
        feed = _json.loads(resp.read().decode())

    VULN_HASHES = set()
    for driver in feed:
        for sample in driver.get("KnownVulnerableSamples", []):
            for field in ("SHA256", "Sha256", "sha256"):
                h = sample.get(field, "").strip().lower()
                if len(h) == 64:
                    VULN_HASHES.add(h)

    print(f'\u2705  Loaded {len(VULN_HASHES)} SHA256 hashes from LOLDrivers feed.')
except Exception as e:
    print(f'\u26a0\ufe0f  LOLDrivers feed unreachable ({e}). Using static fallback.')
    VULN_HASHES = {
        "e2e6e0b6db9428e5c8cd1a0a6f63b93c7e0ecaa4e3b2b0a0d03dedc70a24dc62",  # iqvw64.sys
        "0f16a43f7989034641fd2de3eb268bf2a10a8f0b43430220ac3aa0e547709d1f",  # zamguard64.sys
        "6c69b5c88c882e46de56e0b19e985db3a5e04a99823c8efbf0ad25d5d9f4f8d5",  # aswArPot.sys
        "b45c6136e5e3f9be6c3dbc27b2424b6d8a0b5ba7be42e2c15a4fa77d4b80caff",  # mhyprot2.sys
        "c8f9e1ad7b8ae0fbbd1f6d9be5d48a9b2a7bdae37f2c9e1e33b2f9ab84c0e6f3",  # truesight.sys
    }

print(f'   Hash list ready: {len(VULN_HASHES)} entries.')
````

````python
# ── XSIAM — Detection #2 ───────────────────────────────────────────────────────
# Build an XQL OR-chain filter from the hash set (cap at 200 to stay within query limits)
hash_filter = ' or '.join(
    f'action_module_sha256 = "{h}"' for h in sorted(VULN_HASHES)[:200]
)
QUERY = f'''
dataset = xdr_data
| filter event_type = DRIVER
| filter {hash_filter}
| fields _time, agent_hostname, action_module_path,
         action_module_sha256, action_module_signer
| sort desc _time
'''
run_xsiam(QUERY, label='Det-2 Vuln Driver Hash [XSIAM]')
````

````python
# ── CrowdStrike NG-SIEM — Detection #2 ─────────────────────────────────────────
# LogScale in-values filter
hash_vals = "\n    ".join(f'"{h}"' for h in sorted(VULN_HASHES)[:200])
QUERY = f'''
#event_simpleName=DriverLoad
| SHA256HashData in values=[
    {hash_vals}
  ]
| groupBy([ComputerName, ImageFileName, SHA256HashData], function=count())
| table([_count, ComputerName, ImageFileName, SHA256HashData])
'''
run_crowdstrike(QUERY, label='Det-2 Vuln Driver Hash [CrowdStrike]')
````

````python
# ── Elastic — Detection #2 ──────────────────────────────────────────────────────
# Build a reasonable-sized Lucene OR clause (cap at 50 for query URL safety)
hash_terms = " OR ".join(f"*{h}*" for h in sorted(VULN_HASHES)[:50])
QUERY = f'event.code:"6" AND winlog.event_data.Hashes:({hash_terms})'
run_elastic(QUERY, label='Det-2 Vuln Driver Hash [Elastic]')
````

````python
# ── Microsoft Sentinel — Detection #2 ──────────────────────────────────────────
# Inject hash list into KQL dynamic() array
hash_kql = '", "'.join(sorted(VULN_HASHES)[:200])
QUERY = f'''
let VulnDriverHashes = dynamic(["{hash_kql}"]);
SysmonEvent
| where EventID == 6
| where Hashes has_any (VulnDriverHashes)
| project TimeGenerated, Computer, ImageLoaded, Hashes, Signed, Signature, SignatureStatus
| order by TimeGenerated desc
'''
run_sentinel(QUERY, label='Det-2 Vuln Driver Hash [Sentinel]')
````

## 🔍 Detection #3: KDMapper & Direct Kernel Mapping Tools

KDMapper bypasses the SCM entirely — no Event 7045. Detection must focus on the loader process itself and the transient `iqvw64.sys` that KDMapper drops and quickly removes.

````python
# ── XSIAM — Detection #3 ───────────────────────────────────────────────────────
QUERY = '''
dataset = xdr_data
| filter event_type = PROCESS and event_sub_type = PROCESS_START
| filter action_process_image_name ~= "(?i)(kdmapper|kdu|drvmap|physmem|gdrv-loader|dbk64|kdstealthmap)"
    or action_process_command_line ~= "(?i)(kdmapper|iqvw64|nal\\.sys)"
| fields _time, agent_hostname, actor_effective_username,
         action_process_image_name, action_process_image_path,
         action_process_command_line, actor_process_image_path
| sort desc _time
'''
run_xsiam(QUERY, label='Det-3 KDMapper [XSIAM]')
````

````python
# ── CrowdStrike NG-SIEM — Detection #3 ─────────────────────────────────────────
QUERY = '''
#event_simpleName=ProcessRollup2
| ImageFileName = /(?i)(kdmapper|kdu|drvmap|physmem|gdrv|dbk64|kdstealthmap)/
    OR CommandLine = /(?i)(kdmapper|iqvw64|nal\.sys)/
| table([timestamp, ComputerName, UserName, ImageFileName,
         CommandLine, ParentBaseFileName, SHA256HashData])
| sort(timestamp, order=desc)
'''
run_crowdstrike(QUERY, label='Det-3 KDMapper [CrowdStrike]')
````

````python
# ── Elastic — Detection #3 ──────────────────────────────────────────────────────
QUERY = (
    'event.code:("4688" OR "1") AND ('
    '  process.name:(kdmapper.exe OR kdu.exe OR drvmap.exe OR physmem.exe OR dbk64.exe) OR '
    '  process.command_line:(*kdmapper* OR *iqvw64* OR *nal.sys*)'
    ')'
)
run_elastic(QUERY, label='Det-3 KDMapper [Elastic]')
````

````python
# ── Microsoft Sentinel — Detection #3 ──────────────────────────────────────────
QUERY = '''
let KDMapperNames = dynamic(["kdmapper", "kdu", "drvmap", "physmem",
                             "gdrv-loader", "dbk64", "kdstealthmap"]);
union SecurityEvent, SysmonEvent
| where EventID in (4688, 1)
| extend ProcName = iff(EventID == 4688, NewProcessName, Image)
| extend CmdLine  = iff(EventID == 4688, CommandLine, CommandLine)
| where ProcName has_any (KDMapperNames)
    or CmdLine has_any (KDMapperNames)
    or CmdLine has_any ("iqvw64", "nal.sys")
| project TimeGenerated, Computer, Account, ProcName, CmdLine, ParentProcessName, EventID
| order by TimeGenerated desc
'''
run_sentinel(QUERY, label='Det-3 KDMapper [Sentinel]')
````

## 🔍 Detection #4: Driver Signature Enforcement (DSE) Tampering

Attackers disable DSE by modifying kernel variables (`g_CiEnabled` / `gCiOptions`) in memory, or by setting registry keys under `HKLM\SYSTEM\CurrentControlSet\Control\CI`. Some also use `bcdedit.exe /set testsigning on` to persist DSE-off across reboots.

````python
# ── XSIAM — Detection #4a: CI Registry Tamper ───────────────────────────────────
QUERY_REG = '''
dataset = xdr_data
| filter event_type = REGISTRY
| filter action_registry_key_name ~= "(?i)CurrentControlSet\\\\Control\\\\CI"
| fields _time, agent_hostname, actor_effective_username,
         action_registry_key_name, action_registry_value_name,
         action_registry_data, actor_process_image_path
| sort desc _time
'''
run_xsiam(QUERY_REG, label='Det-4a CI Registry [XSIAM]')

# ── XSIAM — Detection #4b: bcdedit Test-Signing ─────────────────────────────────
QUERY_BCD = '''
dataset = xdr_data
| filter event_type = PROCESS and event_sub_type = PROCESS_START
| filter action_process_command_line ~= "(?i)bcdedit.*(testsigning|bootdebug|nointegritychecks)"
| fields _time, agent_hostname, actor_effective_username,
         action_process_command_line, actor_process_image_path
| sort desc _time
'''
run_xsiam(QUERY_BCD, label='Det-4b bcdedit [XSIAM]')
````

````python
# ── CrowdStrike NG-SIEM — Detection #4 ─────────────────────────────────────────
QUERY = '''
(#event_simpleName=RegSetValue AND RegObjectName = /(?i)\\Control\\CI/)
OR
(#event_simpleName=ProcessRollup2 AND BaseFileName = "bcdedit.exe"
  AND CommandLine = /(?i)(testsigning|bootdebug|nointegritychecks)/)
| table([timestamp, ComputerName, UserName, RegObjectName, RegValueName,
         RegData, CommandLine, ImageFileName])
| sort(timestamp, order=desc)
'''
run_crowdstrike(QUERY, label='Det-4 DSE Tamper [CrowdStrike]')
````

````python
# ── Elastic — Detection #4 ──────────────────────────────────────────────────────
print("--- Part A: CI Registry Tamper (Sysmon Event 13) ---")
run_elastic(
    'event.code:"13" AND registry.path:*\\\\Control\\\\CI*',
    label='Det-4a CI Registry [Elastic]',
)

print("\n--- Part B: bcdedit Test-Signing ---")
run_elastic(
    'event.code:("4688" OR "1") AND process.name:bcdedit.exe AND '
    'process.args:(testsigning OR bootdebug OR nointegritychecks)',
    label='Det-4b bcdedit [Elastic]',
)
````

````python
# ── Microsoft Sentinel — Detection #4 ──────────────────────────────────────────
QUERY = '''
let CiTamper = SysmonEvent
    | where EventID == 13
    | where TargetObject has @"SYSTEM\\CurrentControlSet\\Control\\CI"
    | project TimeGenerated, Computer, User, TargetObject, Details, Image, Detection = "CI Registry Tamper";

let BcdeditTamper = union SecurityEvent, SysmonEvent
    | where EventID in (4688, 1)
    | extend ProcName = iff(EventID == 4688, NewProcessName, Image)
    | extend CmdLine  = iff(EventID == 4688, CommandLine, CommandLine)
    | where ProcName endswith "bcdedit.exe"
    | where CmdLine has_any ("testsigning", "bootdebug", "nointegritychecks")
    | project TimeGenerated, Computer, User = Account, TargetObject = "bcdedit CLI",
              Details = CmdLine, Image = ProcName, Detection = "bcdedit DSE Disable";

union CiTamper, BcdeditTamper
| order by TimeGenerated desc
'''
run_sentinel(QUERY, label='Det-4 DSE Tamper [Sentinel]')
````

## 🔍 Detection #5: Unsigned or Invalidly Signed Driver Load

Sysmon Event 6 captures every driver load with signature metadata. Any unsigned driver loading outside `System32` in a production environment warrants investigation.

> **Tuning note**: Build a baseline of known-good unsigned/legacy drivers first.

````python
# ── XSIAM — Detection #5 ───────────────────────────────────────────────────────
QUERY = '''
dataset = xdr_data
| filter event_type = DRIVER
| filter actor_process_signature_status != "SIGNATURE_SIGNED"
    or actor_process_signature_status = null
| filter action_module_path not ~= "(?i)(\\\\Windows\\\\System32\\\\|\\\\Windows\\\\SysWOW64\\\\|\\\\WinSxS\\\\)"
| fields _time, agent_hostname, action_module_path,
         action_module_sha256, actor_process_signature_status, action_module_signer
| sort desc _time
'''
run_xsiam(QUERY, label='Det-5 Unsigned Driver [XSIAM]')
````

````python
# ── CrowdStrike NG-SIEM — Detection #5 ─────────────────────────────────────────
QUERY = '''
#event_simpleName=DriverLoad
| test(Signed != "true")
| ImageFileName != /(Windows\\System32|SysWOW64|WinSxS)/i
| table([timestamp, ComputerName, UserName, ImageFileName,
         Signed, Signature, SignatureStatus, SHA256HashData])
| sort(timestamp, order=desc)
'''
run_crowdstrike(QUERY, label='Det-5 Unsigned Driver [CrowdStrike]')
````

````python
# ── Elastic — Detection #5 ──────────────────────────────────────────────────────
QUERY = (
    'event.code:"6" AND '
    'NOT winlog.event_data.Signed:"true" AND '
    'NOT winlog.event_data.ImageLoaded:('
    '  "*\\\\Windows\\\\System32\\\\*" OR '
    '  "*\\\\Windows\\\\SysWOW64\\\\*" OR '
    '  "*\\\\WinSxS\\\\*"'
    ')'
)
run_elastic(QUERY, label='Det-5 Unsigned Driver [Elastic]')
````

````python
# ── Microsoft Sentinel — Detection #5 ──────────────────────────────────────────
QUERY = '''
SysmonEvent
| where EventID == 6
| where Signed != "true" or SignatureStatus !in ("Valid")
| where ImageLoaded !startswith @"C:\\Windows\\System32\\"
| where ImageLoaded !startswith @"C:\\Windows\\SysWOW64\\"
| where ImageLoaded !startswith @"C:\\Windows\\WinSxS\\"
| summarize LoadCount = count(), Computers = make_set(Computer), LastSeen = max(TimeGenerated)
    by ImageLoaded, Signed, SignatureStatus, Hashes
| order by LastSeen desc
'''
run_sentinel(QUERY, label='Det-5 Unsigned Driver [Sentinel]')
````

## 🔍 Detection #6: EDR Process Termination After Driver Load

A driver loads → within 5 minutes a known security product process is killed. This sequence is almost never legitimate.

**Monitored security processes:**
```
MsMpEng.exe, MsSense.exe, CSFalconService.exe, SentinelAgent.exe,
SentinelServiceHost.exe, xagt.exe, cb.exe, CylanceSvc.exe, ekrn.exe, avp.exe
```

````python
# ── XSIAM — Detection #6: Driver Load → EDR Kill (5-min window) ─────────────────
QUERY = '''
dataset = xdr_data
| filter event_type = DRIVER
| fields _time as drv_time, agent_hostname, action_module_path as driver_path

| join type=inner (
    dataset = xdr_data
    | filter event_type = PROCESS and event_sub_type = PROCESS_TERMINATE
    | filter action_process_image_name ~= "(?i)(MsMpEng|MsSense|CSFalcon|SentinelAgent|SentinelService|xagt|CylanceSvc|ekrn|avp|cb\\.exe)"
    | fields _time as kill_time, agent_hostname as kill_host,
             action_process_image_name as killed_proc
) as kills on agent_hostname = kill_host

| filter kill_time > drv_time and kill_time < date_add("MINUTE", 5, drv_time)
| fields drv_time, agent_hostname, driver_path, kill_time, killed_proc
| sort desc drv_time
'''
run_xsiam(QUERY, label='Det-6 Driver→EDR Kill [XSIAM]')
````

````python
# ── CrowdStrike NG-SIEM — Detection #6 ─────────────────────────────────────────
# Flag security-product process terminations that occur in bursts
QUERY = '''
#event_simpleName=ProcessRollup2
| ImageFileName = /(?i)(MsMpEng|MsSense|CSFalcon|SentinelAgent|SentinelServiceHost|xagt|CylanceSvc|ekrn|avp)/
| groupBy([ComputerName, ImageFileName],
    function=[count(as=TermCount), min(timestamp, as=First), max(timestamp, as=Last)])
| TermCount > 1
| table([ComputerName, ImageFileName, TermCount, First, Last])
| sort(TermCount, order=desc)
'''
run_crowdstrike(QUERY, label='Det-6 EDR Kill Burst [CrowdStrike]')
````

````python
# ── Elastic — Detection #6: EQL sequence (driver load → EDR process end) ────────
es = _get_es()
eql_query = {
    "query": """
    sequence by host.name with maxspan=5m
      [any where event.code == "6"]
      [process where event.action == "end"
       and process.name : (
           "MsMpEng.exe", "MsSense.exe", "CSFalconService.exe",
           "SentinelAgent.exe", "SentinelServiceHost.exe",
           "xagt.exe", "CylanceSvc.exe", "ekrn.exe", "avp.exe"
       )
      ]
    """,
    "size": 20,
}
try:
    resp = es.eql.search(index=ES_INDEX_PATTERN, body=eql_query)
    seqs = resp.get('hits', {}).get('sequences', [])
    if seqs:
        print(f'\n\U0001f6a8  [Det-6 Elastic]  {len(seqs)} sequence(s) found!\n')
        for i, seq in enumerate(seqs[:10], 1):
            print(f'  Sequence {i}:')
            for ev in seq.get('events', []):
                src = ev.get('_source', {})
                ts  = src.get('@timestamp', '?')
                ec  = src.get('event', {}).get('code', '?')
                hn  = src.get('host', {}).get('name', '?')
                pn  = (src.get('process', {}).get('name', '') or
                       src.get('winlog', {}).get('event_data', {}).get('ImageLoaded', '?'))
                print(f'    [Event {ec}] {ts}  host={hn}  {pn}')
            print()
    else:
        print('\n\u2705  [Det-6 Elastic]  No sequences found in lookback window.')
except Exception as e:
    print(f'\u274c  EQL search failed: {e}')
````

````python
# ── Microsoft Sentinel — Detection #6 ──────────────────────────────────────────
QUERY = '''
let SecurityProcs = dynamic([
    "MsMpEng.exe", "MsSense.exe", "CSFalconService.exe",
    "SentinelAgent.exe", "SentinelServiceHost.exe",
    "xagt.exe", "CylanceSvc.exe", "ekrn.exe", "avp.exe"
]);
let DriverLoads = SysmonEvent
    | where EventID == 6
    | project DriverLoadTime = TimeGenerated, Computer, DriverPath = ImageLoaded;
let ProcessKills = SysmonEvent
    | where EventID == 5
    | extend KilledProc = extract(@"([^\\]+)$", 1, Image)
    | where KilledProc has_any (SecurityProcs)
    | project KillTime = TimeGenerated, Computer, KilledProc;
DriverLoads
| join kind=inner ProcessKills on Computer
| where KillTime between (DriverLoadTime .. (DriverLoadTime + 5m))
| project DriverLoadTime, Computer, DriverPath, KillTime, KilledProc
| order by DriverLoadTime desc
'''
run_sentinel(QUERY, label='Det-6 Driver→EDR Kill [Sentinel]')
````

## 🔍 Detection #7: sc.exe Creating a Kernel Service

Many BYOVD implementations (Terminator, EDRKillShifter, and manual approaches) still use `sc.exe` with `type= kernel` to register the vulnerable driver. The combination of `create`/`config` + `type=kernel` from a non-standard path is very low-noise.

````python
# ── XSIAM — Detection #7 ───────────────────────────────────────────────────────
QUERY = '''
dataset = xdr_data
| filter event_type = PROCESS and event_sub_type = PROCESS_START
| filter action_process_image_name = "sc.exe"
| filter action_process_command_line ~= "(?i)(create|config)"
    and action_process_command_line ~= "(?i)type=?\\s*(kernel|kerneldriver)"
| fields _time, agent_hostname, actor_effective_username,
         action_process_command_line, actor_process_image_name as parent_proc
| sort desc _time
'''
run_xsiam(QUERY, label='Det-7 sc.exe Kernel Service [XSIAM]')
````

````python
# ── CrowdStrike NG-SIEM — Detection #7 ─────────────────────────────────────────
QUERY = '''
#event_simpleName=ProcessRollup2
| BaseFileName = "sc.exe"
| CommandLine = /(?i)(create|config).*type=?\s*(kernel)/
| table([timestamp, ComputerName, UserName, CommandLine, ParentBaseFileName])
| sort(timestamp, order=desc)
'''
run_crowdstrike(QUERY, label='Det-7 sc.exe Kernel Service [CrowdStrike]')
````

````python
# ── Elastic — Detection #7 ──────────────────────────────────────────────────────
QUERY = (
    'event.code:("4688" OR "1") AND '
    'process.name:sc.exe AND '
    'process.command_line:(*create* OR *config*) AND '
    'process.command_line:*kernel*'
)
run_elastic(QUERY, label='Det-7 sc.exe Kernel Service [Elastic]')
````

````python
# ── Microsoft Sentinel — Detection #7 ──────────────────────────────────────────
QUERY = '''
union SecurityEvent, SysmonEvent
| where EventID in (4688, 1)
| extend ProcName = iff(EventID == 4688, NewProcessName, Image)
| extend CmdLine  = iff(EventID == 4688, CommandLine, CommandLine)
| where ProcName endswith "\\sc.exe"
| where CmdLine has_any ("create", "config")
| where CmdLine has "kernel"
| project TimeGenerated, Computer, Account, CmdLine, ParentProcessName
| order by TimeGenerated desc
'''
run_sentinel(QUERY, label='Det-7 sc.exe Kernel Service [Sentinel]')
````

## 📊 Hunting Queries

### Hunt #1: 30-Day Kernel Service Baseline — Rare Installs (Sentinel)

Build a baseline of all kernel driver installs. Sort ascending — the rarest drivers are most suspicious.

````python
QUERY = '''
Event
| where TimeGenerated > ago(30d)
| where EventID == 7045
| extend ServiceType = extract(@"ServiceType\\s*:\\s*([\\w ]+)",  1, EventData)
| extend ImagePath   = extract(@"ImagePath\\s*:\\s*([^\\r\\n]+)",  1, EventData)
| extend ServiceName = extract(@"ServiceName\\s*:\\s*([^\\r\\n]+)", 1, EventData)
| where ServiceType has "kernel"
| summarize InstallCount = count(), FirstSeen = min(TimeGenerated),
            LastSeen = max(TimeGenerated), HostsAffected = dcount(Computer)
    by ServiceName, ImagePath
| order by InstallCount asc
'''
run_sentinel(QUERY, label='Hunt-1 Kernel Service Baseline [Sentinel]')
````

### Hunt #2: Rare Unsigned Driver Loads — Fleet Frequency (Sentinel)

Drivers seen on ≤3 hosts in 7 days that are unsigned or invalid. Widespread legitimate drivers appear on many hosts; BYOVD drivers are targeted and rare.

````python
QUERY = '''
SysmonEvent
| where EventID == 6
| where TimeGenerated > ago(7d)
| summarize HostCount = dcount(Computer), LoadCount = count(),
            Signed = max(Signed), SigStatus = max(SignatureStatus)
    by DriverName = tostring(split(ImageLoaded, "\\")[-1]), Hashes
| where HostCount <= 3
| where Signed != "true" or SigStatus != "Valid"
| order by HostCount asc, LoadCount asc
'''
run_sentinel(QUERY, label='Hunt-2 Rare Unsigned Drivers [Sentinel]')
````

### Hunt #3: .sys Files Written to Non-Standard Paths (XSIAM)

````python
QUERY = '''
dataset = xdr_data
| filter event_type = FILE and event_sub_type in (FILE_WRITE, FILE_CREATE)
| filter action_file_name ~= "\\.sys$"
| filter action_file_path not ~= "(?i)(Windows\\\\System32\\\\|Windows\\\\SysWOW64\\\\|WinSxS\\\\|DriverStore\\\\)"
| fields _time, agent_hostname, actor_effective_username,
         action_file_path, action_file_name, actor_process_image_path
| sort desc _time
'''
run_xsiam(QUERY, label='Hunt-3 .sys Write Outside System32 [XSIAM]')
````

### Hunt #4: LOLDrivers Filename Sweep (CrowdStrike NG-SIEM)

````python
QUERY = '''
#event_simpleName=DriverLoad
| ImageFileName = /(?i)(iqvw64|zamguard|zam64|aswArPot|mhyprot2|truesight|gdrv|ene\.sys|
                        dbutil|procexp152|RTCore64|RTKVHD64|cpuz|msio64|kprocesshacker|
                        ktmw32|nscm|bs_drv|bsdriver|ntbios|amifldrv|physmem|winio|winring)/
| table([timestamp, ComputerName, UserName, ImageFileName, SHA256HashData, Signed, Signature])
| sort(timestamp, order=desc)
'''
run_crowdstrike(QUERY, label='Hunt-4 LOLDrivers Filename Sweep [CrowdStrike]')
````

### Hunt #5: Full BYOVD Attack Chain — .sys Write → Kernel Service → Driver Load (Elastic EQL)

````python
es = _get_es()
eql_query = {
    "query": """
    sequence by host.name with maxspan=10m
      [file where event.action in (\"creation\", \"modification\")
       and file.extension == \"sys\"
       and not file.path : (\"*\\\\System32\\\\*\", \"*\\\\SysWOW64\\\\*\",
                            \"*\\\\WinSxS\\\\*\", \"*\\\\DriverStore\\\\*\")]
      [process where event.action == \"start\"
       and process.name == \"sc.exe\"
       and process.command_line : (\"*type= kernel*\", \"*type=kernel*\")]
      [any where event.code == "6"]
    """,
    "size": 20,
}
try:
    resp = es.eql.search(index=ES_INDEX_PATTERN, body=eql_query)
    seqs = resp.get('hits', {}).get('sequences', [])
    if seqs:
        print(f'\n\U0001f6a8  [Hunt-5 BYOVD Chain Elastic]  {len(seqs)} sequence(s) found!\n')
        for i, seq in enumerate(seqs[:10], 1):
            print(f'  Sequence {i} — {len(seq.get("events", []))} events:')
            for ev in seq.get('events', []):
                src = ev.get('_source', {})
                ts  = src.get('@timestamp', '?')
                ec  = src.get('event', {}).get('code', '?')
                hn  = src.get('host', {}).get('name', '?')
                pn  = (src.get('process', {}).get('name', '') or
                       src.get('file', {}).get('path', '') or
                       src.get('winlog', {}).get('event_data', {}).get('ImageLoaded', '?'))
                print(f'    [Event {ec}] {ts}  host={hn}  {pn}')
            print()
    else:
        print('\n\u2705  [Hunt-5 BYOVD Chain Elastic]  No multi-stage sequences found.')
except Exception as e:
    print(f'\u274c  EQL search failed: {e}')
````

## 🛡️ Mitigation Recommendations

### 1. Enable HVCI (Hypervisor-Protected Code Integrity)
HVCI prevents kernel memory from being writable by non-verified code — eliminating the core primitive BYOVD exploits.
```powershell
Get-CimInstance -ClassName Win32_DeviceGuard -Namespace root\Microsoft\Windows\DeviceGuard
# HypervisorEnforcedCodeIntegrityStatus = 2  →  HVCI active
```

### 2. Deploy the Microsoft WDAC Vulnerable Driver Blocklist
```powershell
# https://learn.microsoft.com/en-us/windows/security/application-security/application-control/
#        windows-defender-application-control/design/microsoft-recommended-driver-block-rules
ConvertFrom-CIPolicy -XmlFilePath .\DriverBlock.xml -BinaryFilePath .\DriverBlock.p7b
```

### 3. Automate LOLDrivers Feed Ingestion
```bash
curl -s https://www.loldrivers.io/api/drivers.json | \\
  jq -r '.[].KnownVulnerableSamples[]?.SHA256 // empty' | sort -u > vuln_hashes.txt
```

### 4. Enable Sysmon Event 6 with Hash Enrichment
```xml
<DriverLoad onmatch="exclude">
    <Signature condition="is">Microsoft Windows</Signature>
    <Signature condition="is">Microsoft Corporation</Signature>
</DriverLoad>
```

### 5. Restrict sc.exe via WDAC / AppLocker
Alert on `sc.exe type=kernel create` from non-approved processes or paths.

### 6. Monthly Driver Fleet Audit
Inventory all running drivers with hashes and compare against LOLDrivers on a schedule.
