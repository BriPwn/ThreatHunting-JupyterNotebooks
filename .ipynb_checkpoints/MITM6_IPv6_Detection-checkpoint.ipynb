# Detecting MITM6 IPv6 Poisoning Attacks

**MITM6** is a DHCPv6-based man-in-the-middle attack tool that exploits Windows' default IPv6 stack behavior to poison DNS and relay authentication. This notebook focuses on detecting the initial IPv6 address assignment phase via **Event ID 51039**.

## The Attack Process

1. **Network Reconnaissance**: Attacker positions themselves on the local network segment

2. **DHCPv6 Poisoning**: MITM6 responds to Windows IPv6 DHCP solicitations with rogue DHCPv6 advertisements
   - Assigns attacker-controlled IPv6 address to victim
   - Sets attacker's host as the DNS server for IPv6 queries

3. **DNS Poisoning**: When victim queries DNS over IPv6:
   - **CNAME Abuse**: Attacker returns CNAME records pointing to relay target (e.g., ADCS server)
   - Victim constructs Kerberos TGS request using CNAME hostname as SPN
   - Enables cross-protocol relay attacks (SMB, HTTP, LDAP)

4. **Authentication Relay**: Captured Kerberos authentication relayed to target services:
   - AD CS (ESC8 attacks for certificate-based persistence)
   - SMB shares for lateral movement
   - HTTP services for further exploitation

5. **Privilege Escalation**: Relayed authentication abused for:
   - Certificate enrollment (ESC8)
   - Resource-based constrained delegation (RBCD)
   - Account takeover via NTLM/Kerberos relay

## Why It's Effective

- **IPv6 Enabled by Default**: Windows prefers IPv6 over IPv4, even if not actively used
- **DHCPv6 Stateless Configuration**: No authorization required for DHCPv6 responses
- **DNS Over IPv6**: Victims automatically query attacker-controlled DNS server
- **Kerberos CNAME Flaw (CVE-2026-20929)**: Windows follows CNAME records and requests TGS for aliased hostname
- **Works Against User Accounts**: Unlike previous techniques limited to machine accounts
- **Cross-Protocol Relay**: Single MITM position enables relay to multiple protocols
- **Low Visibility**: Most organizations don't monitor DHCPv6 or Event ID 51039
- **Bypasses Traditional Defenses**: ARP inspection, DHCP snooping ineffective against IPv6

## Detection Indicators

Primary detection focuses on **Event ID 51039** from the DHCPv6-Client source:

### Event ID 51039: IPv6 Address Assignment

**Source**: Microsoft-Windows-DHCPv6-Client  
**GUID**: {6a1f2b00-6a90-4c38-95a5-5cab3b056778}  
**Channel**: Microsoft-Windows-Dhcpv6-Client/Operational  
**Level**: 4 (Informational)  

**Key Event Data Fields**:
- **Address**: IPv6 address in hexadecimal format (e.g., FE800000000000000192016801380162)
- **InterfaceId**: Numeric network interface identifier
- **StatusCode**: Operation status (0 = success)
- **Frequency**: Rapid IPv6 assignments may indicate MITM6 activity
- **Correlation**: Link with DNS queries, Kerberos authentication events

### Behavioral Indicators

- Multiple hosts receiving IPv6 addresses simultaneously
- IPv6 addresses in unexpected ranges (not matching legitimate DHCPv6 scope)
- Hosts that previously had no IPv6 suddenly getting addresses
- IPv6 DNS server pointing to non-Domain Controller IP
- Spike in Kerberos ticket requests (Event ID 4768, 4769) after IPv6 assignment
- DNS queries over IPv6 to unfamiliar servers

### Example Detection Queries

**Windows Event Log (Event Viewer)**:
```xml
<QueryList>
  <Query Id="0" Path="Microsoft-Windows-Dhcpv6-Client/Operational">
    <Select Path="Microsoft-Windows-Dhcpv6-Client/Operational">
      *[System[Provider[@Name='Microsoft-Windows-DHCPv6-Client'] and EventID=51039]]
    </Select>
  </Query>
</QueryList>
```

**Elastic/Splunk**:
```kql
event.code:51039 AND winlog.provider_name:"Microsoft-Windows-DHCPv6-Client"
```

## Common Tools & Techniques

- **MITM6** - Original DHCPv6 poisoning and relay tool
- **MITM6-Kerberos-CNAME-Abuse** - Enhanced version with CNAME abuse for Kerberos relay
- **krbrelayx.py** - Kerberos relay tool (often used with MITM6)
- **ntlmrelayx.py** - NTLM relay tool from Impacket suite
- **Responder** - Can be combined with MITM6 for comprehensive network attacks

## Attack Chain Correlation

To validate MITM6 activity, correlate Event ID 51039 with:

1. **Event ID 4768** (Kerberos TGT Request) - User requests TGT after IPv6 assignment
2. **Event ID 4769** (Kerberos Service Ticket Request) - TGS request for CNAME target
3. **Event ID 5156** (Windows Filtering Platform Connection) - IPv6 network connections
4. **Sysmon Event ID 22** (DNS Query) - DNS queries over IPv6 to unusual servers
5. **Event ID 4624/4625** (Logon) - Authentication events following IPv6 assignment

- **T1212** - Exploitation for Credential Access

## Mitigation Strategies

1. **Disable IPv6 if Not Needed**: 
   - Group Policy: `Computer Configuration > Administrative Templates > Network > TCPIP Settings > IPv6 Transition Technologies`
   - Registry: `HKLM\SYSTEM\CurrentControlSet\Services\Tcpip6\Parameters\DisabledComponents = 0xFF`

2. **Enable DHCPv6 Guard**: Configure switches to block rogue DHCPv6 advertisements

3. **RA Guard**: Prevent rogue Router Advertisement messages on switch ports

4. **Apply CVE-2026-20929 Patch**: Microsoft's mitigation for Kerberos CNAME abuse
   - Enforces Channel Binding Token (CBT) on HTTPS
   - Enables signing by default for affected protocols

5. **Monitor Event ID 51039**: Baseline normal IPv6 assignments and alert on deviations

6. **Network Segmentation**: Limit Layer 2 broadcast domains to reduce MITM attack surface

7. **Enable SMB Signing**: Prevent SMB relay attacks
   - GPO: `Computer Configuration > Policies > Windows Settings > Security Settings > Local Policies > Security Options`
   - Set: `Microsoft network server: Digitally sign communications (always) = Enabled`

8. **LDAP Signing and Channel Binding**: Prevent LDAP relay

9. **Deploy 802.1X**: Port-based network access control reduces rogue device connections

10. **EDR/NDR Solutions**: Deploy endpoint and network detection for DHCPv6 anomalies

### Return Elastic Results - MITM6 IPv6 Detection

````python
import json
import warnings
from datetime import datetime, timedelta
from elasticsearch import Elasticsearch
from tabulate import tabulate
from collections import defaultdict
warnings.filterwarnings('ignore')

# Connection details - UPDATE THESE WITH YOUR ENVIRONMENT
ES_ENDPOINT = "https://192.168.138.203:9200"
API_KEY = "<YOUR-API-KEY>"

# Index patterns
SYSTEM_INDEX = "logs-windows.system-*"
SECURITY_INDEX = "logs-windows.forwarded-*"
ALL_LOGS = "logs-*"

# Detection time window
LOOKBACK_HOURS = 72
CORRELATION_WINDOW_MINUTES = 15  # Time window to correlate IPv6 with Kerberos events

try:
    # Initialize Elasticsearch client
    client = Elasticsearch(
        [ES_ENDPOINT],
        api_key=API_KEY,
        verify_certs=False,
        ssl_show_warn=False
    )
    
    # Test connection
    if client.ping():
        print("✓ Successfully connected to Elasticsearch")
        info = client.info()
        print(f"✓ Cluster: {info['cluster_name']}")
        print(f"✓ Version: {info['version']['number']}")
        print(f"✓ Detection Window: Last {LOOKBACK_HOURS} hours\n")
    else:
        print("✗ Could not connect to Elasticsearch")
        exit(1)
    
    # Calculate time range
    time_from = (datetime.utcnow() - timedelta(hours=LOOKBACK_HOURS)).strftime('%Y-%m-%dT%H:%M:%S.000Z')
    
    # ========================================================================
    # STEP 1: Search for Event ID 51039 (IPv6 Address Assignment)
    # ========================================================================
    print("="*80)
    print("STEP 1: Searching for IPv6 Address Assignments (Event ID 51039)")
    print("="*80)
    
    ipv6_query = {
        "query": {
            "bool": {
                "must": [
                    {"term": {"event.code": "51039"}},
                    {"match": {"winlog.provider_name": "Microsoft-Windows-DHCPv6-Client"}},
                    {"range": {"@timestamp": {"gte": time_from}}}
                ]
            }
        },
        "size": 100,
        "sort": [{"@timestamp": {"order": "desc"}}]
    }
    
    ipv6_response = client.search(index=ALL_LOGS, body=ipv6_query)
    ipv6_count = ipv6_response['hits']['total']['value']
    
    print(f"\nFound {ipv6_count} IPv6 address assignment events\n")
    
    if ipv6_count == 0:
        print("✓ No IPv6 assignments detected in the time window.")
        print("  Note: This could mean:")
        print("    - No MITM6 activity (good!)")
        print("    - IPv6 is disabled on monitored systems")
        print("    - Event ID 51039 is not being collected")
    else:
        print("⚠️  IPv6 address assignments detected. Analyzing for MITM6 indicators...\n")
        
        # Display IPv6 events in table format
        ipv6_table = []
        ipv6_hosts = {}  # Store timestamps per host for correlation
        
        for i, hit in enumerate(ipv6_response['hits']['hits'], 1):
            source = hit['_source']
            timestamp = source.get('@timestamp', 'N/A')
            hostname = source.get('host', {}).get('name', 'Unknown')
            
            # Store for correlation
            if hostname not in ipv6_hosts:
                ipv6_hosts[hostname] = []
            ipv6_hosts[hostname].append(timestamp)
            
            row = {
                "#": i,
                "Timestamp": timestamp,
                "Host": hostname,
                "Computer": source.get('winlog', {}).get('computer_name', 'N/A'),
                "Event Message": source.get('message', 'N/A')[:80] + '...' if len(source.get('message', '')) > 80 else source.get('message', 'N/A')
            }
            ipv6_table.append(row)
        
        print("--- Recent IPv6 Address Assignments ---")
        print(tabulate(ipv6_table[:10], headers="keys", tablefmt="grid"))  # Show first 10
        
        # Show sample event detail
        if ipv6_response['hits']['hits']:
            print("\n--- Sample Event ID 51039 Detail ---")
            sample = ipv6_response['hits']['hits'][0]['_source']
            print(f"Timestamp: {sample.get('@timestamp')}")
            print(f"Host: {sample.get('host', {}).get('name')}")
            print(f"Provider: {sample.get('winlog', {}).get('provider_name')}")
            print(f"Channel: {sample.get('winlog', {}).get('channel')}")
            print(f"Message: {sample.get('message')}")
            
            event_data = sample.get('winlog', {}).get('event_data', {})
            print(f"\nEvent Data:")
            print(f"  - Address (hex): {event_data.get('Address', 'N/A')}")
            print(f"  - InterfaceId: {event_data.get('InterfaceId', 'N/A')}")
            print(f"  - StatusCode: {event_data.get('StatusCode', 'N/A')}")
            
            # Try to decode hex IPv6 address if present
            if 'Address' in event_data:
                hex_addr = event_data['Address']
                print(f"\n  Note: IPv6 address is in hex format (32 chars = 16 bytes)")
                print(f"  Example: FE800000000000000192016801380162")
                print(f"  Decoded: FE80::0192:0168:0138:0162 (link-local IPv6)")
            
            print(f"\nFull Event Data: {json.dumps(event_data, indent=2)}")
    
    # ========================================================================
    # STEP 2: Check for Multiple IPv6 Assignments (Potential Scanning)
    # ========================================================================
    print("\n" + "="*80)
    print("STEP 2: Analyzing IPv6 Assignment Patterns")
    print("="*80)
    
    if ipv6_count > 0:
        # Count assignments per host
        host_counts = defaultdict(int)
        for hit in ipv6_response['hits']['hits']:
            hostname = hit['_source'].get('host', {}).get('name', 'Unknown')
            host_counts[hostname] += 1
        
        print("\n--- IPv6 Assignments Per Host ---")
        pattern_table = []
        suspicious_hosts = []
        
        for host, count in sorted(host_counts.items(), key=lambda x: x[1], reverse=True):
            severity = "⚠️  SUSPICIOUS" if count > 5 else "ℹ️  Normal" if count <= 2 else "⚡ Elevated"
            if count > 3:
                suspicious_hosts.append(host)
            
            pattern_table.append({
                "Host": host,
                "Assignment Count": count,
                "Severity": severity
            })
        
        print(tabulate(pattern_table[:15], headers="keys", tablefmt="grid"))
        
        if suspicious_hosts:
            print(f"\n⚠️  WARNING: {len(suspicious_hosts)} host(s) with elevated IPv6 assignment counts!")
            print(f"   Hosts: {', '.join(suspicious_hosts[:5])}")
            print("   This may indicate:")
            print("     - Active MITM6 attack")
            print("     - Network instability or DHCP issues")
            print("     - Investigate these hosts immediately")
    
    # ========================================================================
    # STEP 3: Correlate with Kerberos Activity (Event IDs 4768, 4769)
    # ========================================================================
    print("\n" + "="*80)
    print("STEP 3: Correlating IPv6 Assignments with Kerberos Activity")
    print("="*80)
    
    if ipv6_count > 0 and ipv6_hosts:
        print(f"\nSearching for Kerberos events within {CORRELATION_WINDOW_MINUTES} minutes of IPv6 assignments...\n")
        
        # Search for Kerberos ticket requests
        kerberos_query = {
            "query": {
                "bool": {
                    "must": [
                        {"terms": {"event.code": ["4768", "4769"]}},
                        {"range": {"@timestamp": {"gte": time_from}}}
                    ]
                }
            },
            "size": 500,
            "sort": [{"@timestamp": {"order": "desc"}}]
        }
        
        kerb_response = client.search(index=ALL_LOGS, body=kerberos_query)
        kerb_count = kerb_response['hits']['total']['value']
        
        print(f"Found {kerb_count} Kerberos events (4768: TGT, 4769: TGS)")
        
        if kerb_count > 0:
            # Basic correlation: Check if Kerberos events occur on same hosts
            kerb_hosts = set()
            for hit in kerb_response['hits']['hits']:
                hostname = hit['_source'].get('host', {}).get('name', 'Unknown')
                kerb_hosts.add(hostname)
            
            overlapping_hosts = set(ipv6_hosts.keys()) & kerb_hosts
            
            if overlapping_hosts:
                print(f"\n🚨 CRITICAL: {len(overlapping_hosts)} host(s) show both IPv6 assignment AND Kerberos activity!")
                print("\n--- Correlated Hosts (Potential MITM6 Targets) ---")
                
                corr_table = []
                for host in list(overlapping_hosts)[:10]:
                    corr_table.append({
                        "Host": host,
                        "IPv6 Events": len(ipv6_hosts[host]),
                        "Risk": "🚨 HIGH - Investigate Immediately"
                    })
                
                print(tabulate(corr_table, headers="keys", tablefmt="grid"))
                
                print("\n--- IMMEDIATE INVESTIGATION STEPS ---")
                print("1. Isolate affected systems from the network if active attack suspected")
                print("2. Verify IPv6 configuration - check DNS server settings")
                print("3. Review Kerberos authentication logs for unusual SPNs or relay patterns")
                print("4. Check for rogue DHCPv6 servers on network segments")
                print("5. Analyze network traffic for MITM6 tool signatures")
                print("6. Review AD CS certificate enrollment logs (Event ID 4886, 4887)")
                print("7. Verify no unauthorized certificates have been issued")
                print("8. Scan for suspicious PowerShell or Python processes (mitm6, krbrelayx)")
            else:
                print("\n✓ No direct correlation between IPv6 and Kerberos events on same hosts")
                print("  This reduces (but doesn't eliminate) MITM6 suspicion")
        else:
            print("\nℹ️  No Kerberos events found in time window")
            print("   Note: MITM6 attacks typically generate Kerberos activity")
    
    # ========================================================================
    # STEP 4: Network-Wide IPv6 Pattern Analysis
    # ========================================================================
    print("\n" + "="*80)
    print("STEP 4: Network-Wide Pattern Analysis")
    print("="*80)
    
    if ipv6_count > 0:
        # Aggregate by time buckets to detect synchronized attacks
        time_bucket_query = {
            "query": {
                "bool": {
                    "must": [
                        {"term": {"event.code": "51039"}},
                        {"range": {"@timestamp": {"gte": time_from}}}
                    ]
                }
            },
            "aggs": {
                "events_over_time": {
                    "date_histogram": {
                        "field": "@timestamp",
                        "fixed_interval": "1h"
                    }
                }
            },
            "size": 0
        }
        
        time_agg_response = client.search(index=ALL_LOGS, body=time_bucket_query)
        buckets = time_agg_response['aggregations']['events_over_time']['buckets']
        
        if buckets:
            print("\n--- IPv6 Assignment Timeline (Hourly) ---")
            timeline_table = []
            max_count = max([b['doc_count'] for b in buckets]) if buckets else 0
            
            for bucket in buckets[-24:]:  # Last 24 hours
                if bucket['doc_count'] > 0:
                    timestamp = bucket['key_as_string']
                    count = bucket['doc_count']
                    bar = '█' * int((count / max_count) * 20) if max_count > 0 else ''
                    
                    alert = "🚨" if count > 20 else "⚠️" if count > 10 else ""
                    
                    timeline_table.append({
                        "Time": timestamp,
                        "Events": count,
                        "Pattern": bar,
                        "Alert": alert
                    })
            
            if timeline_table:
                print(tabulate(timeline_table, headers="keys", tablefmt="grid"))
                
                # Detect spikes
                spike_threshold = max_count * 0.7 if max_count > 0 else 0
                spikes = [b for b in buckets if b['doc_count'] > spike_threshold and b['doc_count'] > 10]
                
                if spikes:
                    print(f"\n⚠️  {len(spikes)} time period(s) with elevated IPv6 activity detected!")
                    print("   This pattern may indicate:")
                    print("     - Coordinated MITM6 attack across multiple systems")
                    print("     - Mass system startup/network reconnection")
                    print("     - Review affected time periods for correlation with incidents")
    
    # ========================================================================
    # FINAL SUMMARY
    # ========================================================================
    print("\n" + "="*80)
    print("DETECTION SUMMARY")
    print("="*80)
    
    print(f"\n📊 Statistics:")
    print(f"   - Total IPv6 Assignments: {ipv6_count}")
    print(f"   - Unique Hosts: {len(ipv6_hosts) if ipv6_hosts else 0}")
    print(f"   - Time Window: {LOOKBACK_HOURS} hours")
    
    if ipv6_count == 0:
        print("\n✅ No IPv6 assignment activity detected - Low risk")
    elif ipv6_count < 5:
        print("\n✅ Minimal IPv6 activity - Appears normal")
        print("   Continue baseline monitoring")
    elif suspicious_hosts:
        print("\n🚨 POTENTIAL MITM6 ATTACK DETECTED!")
        print("   - Suspicious IPv6 patterns identified")
        print("   - Correlation with Kerberos activity")
        print("   - IMMEDIATE RESPONSE REQUIRED")
    else:
        print("\n⚠️  Elevated IPv6 activity detected")
        print("   - Monitor closely for attack indicators")
        print("   - Verify legitimate business activity")
    
    print("\n🔍 Recommended Actions:")
    print("   1. Review IPv6 configuration across enterprise")
    print("   2. Validate DHCPv6 server legitimacy")
    print("   3. Enable DHCPv6 Guard and RA Guard on switches")
    print("   4. Apply CVE-2026-20929 patches if not already deployed")
    print("   5. Consider disabling IPv6 if not business-critical")
    print("   6. Implement continuous monitoring of Event ID 51039")
    
except Exception as e:
    print(f"❌ Error: {e}")
    import traceback
    traceback.print_exc()
finally:
    if 'client' in locals():
        client.close()
        print("\n✓ Connection closed")
````

## Additional Hunting Queries

### Query 1: Detect IPv6 DNS Server Changes

MITM6 modifies IPv6 DNS server settings to point to attacker-controlled infrastructure.

```powershell
# PowerShell - Check current IPv6 DNS configuration
Get-NetAdapter | Get-DnsClientServerAddress -AddressFamily IPv6

# Look for DNS servers that don't match your Domain Controllers
```

### Query 2: Detect Unusual Kerberos SPNs (CNAME Abuse)

```kql
event.code:4769 AND
winlog.event_data.ServiceName:*
| stats count by winlog.event_data.ServiceName, winlog.event_data.TargetUserName
| where count < 5  # Rare SPNs may indicate CNAME abuse
```

### Query 3: Network Flow Analysis for DHCPv6

```kql
destination.port:547 OR source.port:547
# DHCPv6 uses UDP port 547
# Look for unexpected DHCPv6 servers
```

### Query 4: Sysmon DNS Queries over IPv6

```kql
event.code:22 AND 
dns.question.name:* AND
source.ip:(fe80* OR 2001* OR fc00*)
# Sysmon Event ID 22: DNS Query
# Filter for IPv6 source addresses
```
