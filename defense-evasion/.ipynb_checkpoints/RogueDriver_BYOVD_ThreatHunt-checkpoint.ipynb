## 📋 Overview

This notebook provides threat hunting and detection strategies for **Rogue Driver Installation** via the **Bring Your Own Vulnerable Driver (BYOVD)** technique. BYOVD is one of the most prevalent defense-evasion primitives in modern intrusions — used by ransomware gangs (RansomHub, LockBit, BlackCat), nation-state actors (Scattered Spider, VENOMOUS BEAR/Turla), and commodity toolkits alike.

### How BYOVD Works

Attackers bring a **legitimately signed but vulnerable driver** onto the target, load it as a kernel service, then exploit its vulnerability to gain kernel read/write primitives:

| Goal | Example Action |
|------|---------------|
| Kill EDR/AV | Terminate protected sensor processes |
| Disable DSE | Turn off Driver Signature Enforcement, load unsigned rootkit |
| Remove EDR callbacks | Unregister kernel notification routines |
| Dump LSASS | Read process memory as kernel |

### Real-World Examples

| Driver | CVE | Abused By |
|--------|-----|-----------|
| `iqvw64.sys` (Intel NIC diag) | CVE-2015-2291 | Scattered Spider |
| `zamguard64.sys` / `zam64.sys` | — | BlackCat / Terminator tool |
| `aswArPot.sys` (Avast Anti-Rootkit) | — | Ransomware campaigns (Nov 2024) |
| `mhyprot2.sys` (Genshin Impact anti-cheat) | — | Multiple ransomware groups |
| `truesight.sys` (TrueSight) | — | EDRKillShifter / RansomHub |
| `PROCEXP152.sys` (Process Explorer) | — | Malvertising campaign |

## 🔴 Attack Phases

### Phase 1 — Drop the Vulnerable Driver
```
C:\Users\Public\iqvw64.sys
C:\Windows\Temp\zamguard64.sys
C:\ProgramData\aswArPot.sys
```

### Phase 2 — Load as a Kernel Service
```cmd
sc create VulnDrv binPath= C:\Users\Public\iqvw64.sys type= kernel start= demand
sc start VulnDrv
```
Tools like **KDMapper** bypass the SCM entirely — leaving no Event 7045.

### Phase 3 — Exploit Kernel R/W Primitive
The attacker sends `DeviceIoControl` calls to the loaded driver, obtaining arbitrary kernel read/write.

### Phase 4 — Payload (EDR Kill / DSE Disable / Rootkit)
```
Patch nt!g_CiEnabled = 0              // Disable Code Integrity
ZwTerminateProcess(EDR_PID)           // Kill the EDR sensor
NtLoadDriver(malicious_unsigned.sys)  // Load unsigned rootkit
```

### Key Tooling

| Tool | Purpose |
|------|---------|
| **KDMapper** | Maps drivers into kernel memory, bypasses SCM |
| **Terminator** (spyboy) | BYOVD wrapper targeting 23+ AV/EDR products |
| **EDRKillShifter** (RansomHub) | Automated BYOVD loader via ransomware affiliate program |
| **AuKill** (LockBit) | Custom BYOVD tool based on Backstab |
| **LIGHTSHOW** (UNC2970) | Nation-state BYOVD using ENE Technology driver |

### Step 3 — Initialize Clients & Helper Functions

Run this cell once. All detection cells call these shared helpers.

````python
import hashlib, random, string, time, json, warnings, requests
# FalconPy OAuth2 imported lazily inside _get_cs_oauth2()
from datetime import timedelta
from elasticsearch import Elasticsearch
from azure.identity import ClientSecretCredential
from azure.monitor.query import LogsQueryClient, LogsQueryStatus
from tabulate import tabulate

warnings.filterwarnings('ignore')

# Strip scheme from FQDN so f-string URL construction always works
XSIAM_FQDN = XSIAM_FQDN.removeprefix('https://').removeprefix('http://').rstrip('/')


# ══════════════════════════════════════════════════════════════════════════════
# XSIAM helpers
#
# Standard key  →  x-xdr-token: <api_key>  (plain, no hashing)
# Advanced key  →  Authorization: <SHA-256(api_key + nonce + timestamp_ms)>
#                                  NOTE: raw hex digest, NO 'Bearer'/'Advanced' prefix
# Reference: https://cortex-panw.stoplight.io/docs/cortex-xdr/getting-started-1/l3
# ══════════════════════════════════════════════════════════════════════════════

def _xsiam_headers():
    base = {
        'x-xdr-auth-id': str(XSIAM_API_KEY_ID),
        'Content-Type':  'application/json',
    }
    if XSIAM_AUTH_TYPE.lower() == 'standard':
        # Standard key: pass the key directly as x-xdr-token
        base['x-xdr-token'] = XSIAM_API_KEY

    elif XSIAM_AUTH_TYPE.lower() == 'advanced':
        # Advanced key: SHA-256(api_key + nonce + timestamp_ms)
        # Authorization value is the RAW hex digest — no prefix
        nonce = ''.join(random.choices(string.ascii_letters + string.digits, k=64))
        ts    = str(int(time.time() * 1000))
        h     = hashlib.sha256((XSIAM_API_KEY + nonce + ts).encode()).hexdigest()
        base.update({
            'x-xdr-nonce':     nonce,
            'x-xdr-timestamp': ts,
            'Authorization':   h,   # raw digest only — NOT 'Advanced h' or 'Bearer h'
        })

    else:
        raise ValueError(f"Unknown XSIAM_AUTH_TYPE '{XSIAM_AUTH_TYPE}'. Use 'standard' or 'advanced'.")

    return base


def _xsiam_extract_reply(r, step):
    """Parse an XSIAM HTTP response and return the value of the 'reply' key.

    Handles all observed shapes:
      {"reply": {"execution_id": "abc"}}   -> inner dict
      {"reply": "abc"}                     -> string directly
      {"reply": "{\"status\":\"SUCCESS\"}"}  -> JSON-string, unwrapped
    Raises RuntimeError with the raw body on unexpected structure.
    """
    raw = r.text
    try:
        body = r.json()
    except Exception:
        raise RuntimeError(
            f'XSIAM {step}: not valid JSON.\n'
            f'  HTTP {r.status_code}  body: {raw[:800]}'
        )
    if not isinstance(body, dict):
        raise RuntimeError(
            f'XSIAM {step}: expected JSON object, got {type(body).__name__}.\n'
            f'  HTTP {r.status_code}  body: {raw[:800]}'
        )
    if 'reply' not in body:
        raise RuntimeError(
            f'XSIAM {step}: no "reply" key.  Keys: {list(body.keys())}\n'
            f'  HTTP {r.status_code}  body: {raw[:800]}'
        )
    reply = body['reply']
    if isinstance(reply, str):
        try:
            unwrapped = json.loads(reply)
            if isinstance(unwrapped, (dict, list)):
                reply = unwrapped
        except Exception:
            pass
    return reply


def run_xsiam(query, label='XSIAM'):
    """Submit an XQL query and return result rows as a list of dicts.

    Correct API flow per Cortex XSIAM REST API docs:
      1. POST /public_api/v1/xql/start_xql_query/
         Body: {"request_data": {"query": "...", "timeframe": {...}}}
         Response: {"reply": "<query_id>"}  OR  {"reply": {"execution_id": "<query_id>"}}

      2. POST /public_api/v1/xql/get_query_results/          <- NOTE: different path
         Body: {"request_data": {"query_id": "<id>",         <- NOTE: query_id not execution_id
                                 "pending_flag": true,
                                 "limit": 1000,
                                 "format": "json"}}
         Response: {"reply": {"status": "SUCCESS", "results": {"data": [...]}}}
    """
    # ── Step 1: Start the query ───────────────────────────────────────────────
    start_url = f'https://{XSIAM_FQDN}/public_api/v1/xql/start_xql_query/'
    payload   = {
        'request_data': {
            'query':     query,
            'timeframe': {'relativeTime': XSIAM_LOOKBACK * 60 * 1000},
        }
    }
    r = requests.post(start_url, json=payload, headers=_xsiam_headers(), timeout=30)
    if r.status_code != 200:
        raise RuntimeError(
            f'XSIAM start_xql_query HTTP {r.status_code}.\n'
            f'  Body: {r.text[:600]}'
        )

    reply = _xsiam_extract_reply(r, 'start_xql_query')

    # reply is either the query_id string directly, or a dict containing it
    if isinstance(reply, str):
        query_id = reply
    elif isinstance(reply, dict):
        query_id = reply.get('execution_id') or reply.get('query_id')
        if not query_id:
            raise RuntimeError(
                f'XSIAM: no query_id/execution_id in reply.\n'
                f'  Reply: {json.dumps(reply)[:600]}'
            )
    else:
        raise RuntimeError(f'XSIAM: unexpected reply type {type(reply).__name__}: {reply!r}')

    # ── Step 2: Poll for results ──────────────────────────────────────────────
    # Correct endpoint and field names per API docs
    poll_url = f'https://{XSIAM_FQDN}/public_api/v1/xql/get_query_results/'
    poll_payload = {
        'request_data': {
            'query_id':    query_id,
            'pending_flag': True,
            'limit':        1000,
            'format':       'json',
        }
    }

    # ── Strategy 1: pending_flag=true makes the server block until done ──────
    # Try a single long-timeout request first; many XSIAM versions return
    # results immediately when pending_flag is true rather than requiring polling.
    print(f'  Waiting for results (query_id: {query_id}) ...')
    blocking_payload = {
        'request_data': {
            'query_id':    query_id,
            'pending_flag': True,
            'limit':        1000,
            'format':       'json',
        }
    }
    try:
        r = requests.post(poll_url, json=blocking_payload,
                          headers=_xsiam_headers(), timeout=120)
        if r.status_code == 200:
            reply = _xsiam_extract_reply(r, 'get_query_results (blocking)')
            if isinstance(reply, dict):
                status = (reply.get('status') or reply.get('query_status') or '').upper()
                print(f'  Blocking call status: {status!r}')
                if status == 'SUCCESS':
                    results_block = reply.get('results') or reply
                    rows = results_block.get('data', []) if isinstance(results_block, dict) else []
                    _display(label, rows)
                    return rows
                # If not SUCCESS, fall through to polling loop below
    except requests.exceptions.Timeout:
        print('  Blocking call timed out after 120s — falling back to poll loop.')

    # ── Strategy 2: poll loop with debug output on every attempt ─────────────
    print('  Polling for results...')
    for attempt in range(40):
        time.sleep(5)
        r = requests.post(poll_url, json=poll_payload,
                          headers=_xsiam_headers(), timeout=30)
        if r.status_code != 200:
            ct = r.headers.get('Content-Type', '')
            raise RuntimeError(
                f'XSIAM poll HTTP {r.status_code} (attempt {attempt+1}).\n'
                f'  Content-Type: {ct}\n'
                f'  Body: {r.text[:600]}'
            )

        reply = _xsiam_extract_reply(r, 'get_query_results')

        if not isinstance(reply, dict):
            raise RuntimeError(
                f'XSIAM poll reply is {type(reply).__name__}, expected dict.\n'
                f'  Value: {str(reply)[:600]}  |  Raw: {r.text[:400]}'
            )

        # Print full reply on first 3 attempts so we can see what fields are returned
        if attempt < 3:
            safe = {k: v for k, v in reply.items() if k != 'results'}
            print(f'  Attempt {attempt+1} reply keys/status: {safe}')

        status = (reply.get('status') or reply.get('query_status') or '').upper()
        if status == 'SUCCESS':
            results_block = reply.get('results') or reply
            rows = results_block.get('data', []) if isinstance(results_block, dict) else []
            _display(label, rows)
            return rows
        elif status in ('FAILED', 'CANCELLED', 'TIMEOUT', 'ERROR'):
            raise RuntimeError(
                f'XQL query {status}.\n'
                f'  Reply: {json.dumps(reply)[:600]}'
            )
        # PENDING / RUNNING — keep polling
    raise TimeoutError(
        f'XSIAM query timed out after {40*5}s.\n'
        f'  Last reply (non-results keys): { {k:v for k,v in reply.items() if k!="results"} }'
    )


# ══════════════════════════════════════════════════════════════════════════════
# CrowdStrike NG-SIEM helpers
#
# Strategy: FalconPy OAuth2 for token management only.
# All API calls are made via raw requests to avoid FalconPy's broken
# ngsiem_search_payload() builder (causes HTTP 400 "No content" regardless
# of whether keyword args or search= dict are used).
#
# Endpoints:
#   POST /humio/api/v1/repositories/{repo}/queryjobs       → start job
#   GET  /humio/api/v1/repositories/{repo}/queryjobs/{id}  → poll status/results
#
# Required API scope: Next-Gen SIEM → Falcon Long Term Repository → Read + Write
# ══════════════════════════════════════════════════════════════════════════════
_cs_oauth2 = None   # OAuth2 instance — created lazily, handles token refresh

def _get_cs_oauth2():
    global _cs_oauth2
    if _cs_oauth2 is None:
        from falconpy import OAuth2
        _cs_oauth2 = OAuth2(client_id=CS_CLIENT_ID, client_secret=CS_CLIENT_SECRET)
    return _cs_oauth2

def _cs_bearer_token():
    """Return a valid Bearer token via FalconPy OAuth2 (handles refresh automatically)."""
    result = _get_cs_oauth2().token()
    sc = result.get('status_code', 0)
    if sc not in (200, 201):
        raise RuntimeError(f'CrowdStrike OAuth2 failed (HTTP {sc}): {result.get("body")}')
    return result['body']['access_token']

def _cs_base_url():
    """Return the regional base URL — FalconPy autodiscovers region on first token call.
    
    After OAuth2.token() succeeds, the resolved URL is stored at auth._config.base_url.
    Falls back to US-1 default if not yet resolved.
    """
    auth = _get_cs_oauth2()
    url  = getattr(getattr(auth, '_config', None), 'base_url', None) \
           or getattr(auth, 'base_url', 'https://api.crowdstrike.com')
    return url.rstrip('/')

def run_crowdstrike(query, label='CrowdStrike NG-SIEM'):
    """Submit a LogScale queryjobs search via direct HTTP and return rows.

    Uses FalconPy OAuth2 for token management only. The NGSIEM service class
    payload builder (ngsiem_search_payload) has a serialisation bug that causes
    HTTP 400 "No content was received" regardless of how kwargs are provided,
    so we bypass it entirely with raw requests.

    Flow:
      1. POST /humio/api/v1/repositories/{repo}/queryjobs  → get job id
      2. GET  /humio/api/v1/repositories/{repo}/queryjobs/{id} → poll until done
      3. Return response body 'events' list
    """
    token    = _cs_bearer_token()
    base_url = _cs_base_url()
    repo     = CS_REPOSITORY
    headers  = {
        'Authorization': f'Bearer {token}',
        'Content-Type':  'application/json',
        'Accept':        'application/json',
    }

    # ── Step 1: Submit the queryjob ───────────────────────────────────────────
    start_url = f'{base_url}/humio/api/v1/repositories/{repo}/queryjobs'
    payload   = {
        'queryString': query,
        'start':       CS_LOOKBACK,
        'end':         'now',
        'isLive':      False,
    }
    r = requests.post(start_url, json=payload, headers=headers, timeout=30)
    if r.status_code not in (200, 201):
        raise RuntimeError(
            f'NGSIEM queryjobs POST failed (HTTP {r.status_code}).\n'
            f'  URL: {start_url}\n'
            f'  Body: {r.text[:400]}'
        )
    job_id = r.json().get('id')
    if not job_id:
        raise RuntimeError(f'NGSIEM: no job id in POST response: {r.text[:200]}')

    # ── Step 2: Poll until done ───────────────────────────────────────────────
    poll_url = f'{start_url}/{job_id}'
    for attempt in range(40):
        time.sleep(3)
        p = requests.get(poll_url, headers=headers, timeout=30)
        if p.status_code != 200:
            raise RuntimeError(
                f'NGSIEM queryjobs GET failed (HTTP {p.status_code}, attempt {attempt+1}).\n'
                f'  Body: {p.text[:400]}'
            )
        data = p.json()
        if data.get('done'):
            rows = data.get('events', [])
            _display(label, rows)
            return rows
    raise TimeoutError(
        f'CrowdStrike NG-SIEM query timed out after {40*3}s (job_id={job_id})'
    )


# ══════════════════════════════════════════════════════════════════════════════
# Elastic helpers
# ══════════════════════════════════════════════════════════════════════════════
_es_client = None

def _get_es():
    global _es_client
    if _es_client is None:
        _es_client = Elasticsearch(
            ES_HOST, api_key=ES_API_KEY,
            verify_certs=ES_VERIFY_CERTS, ssl_show_warn=False,
        )
    return _es_client

def run_elastic(lucene_query, label='Elastic', size=50):
    client = _get_es()
    body = {
        'query': {
            'bool': {
                'must': [
                    {'query_string': {'query': lucene_query}},
                    {'range': {'@timestamp': {'gte': f'now-{ES_LOOKBACK_DAYS}d', 'lte': 'now'}}},
                ]
            }
        },
        'size': size,
        'sort': [{'@timestamp': {'order': 'desc'}}],
    }
    resp  = client.search(index=ES_INDEX_PATTERN, body=body)
    total = resp['hits']['total']['value']
    rows  = [h['_source'] for h in resp['hits']['hits']]
    print(f'\n[{label}] Total matches: {total}  (showing up to {size})')
    _display(label, rows)
    return rows


# ══════════════════════════════════════════════════════════════════════════════
def _display(label, rows, max_rows=25, max_col=90):
    if not rows:
        print(f'\n\u2705  [{label}]  No results — no hits in the lookback window.')
        return
    print(f'\n\U0001f6a8  [{label}]  {len(rows)} result(s) found!\n')
    if rows and isinstance(rows[0], dict):
        disp = [{k: str(v)[:max_col] for k, v in row.items()} for row in rows[:max_rows]]
        print(tabulate(disp, headers='keys', tablefmt='grid'))
        if len(rows) > max_rows:
            print(f'  ... ({len(rows) - max_rows} additional rows truncated)')
    else:
        for row in rows[:max_rows]:
            print(row)


print(f'\u2705  Helpers loaded  |  XSIAM auth: {XSIAM_AUTH_TYPE}  |  endpoint: https://{XSIAM_FQDN}')

def _display(label, rows, max_rows=25, max_col=90):
    if not rows:
        print(f'\n\u2705  [{label}]  No results — no hits in the lookback window.')
        return
    print(f'\n\U0001f6a8  [{label}]  {len(rows)} result(s) found!\n')
    if rows and isinstance(rows[0], dict):
        disp = [{k: str(v)[:max_col] for k, v in row.items()} for row in rows[:max_rows]]
        print(tabulate(disp, headers='keys', tablefmt='grid'))
        if len(rows) > max_rows:
            print(f'  ... ({len(rows) - max_rows} additional rows truncated)')
    else:
        for row in rows[:max_rows]:
            print(row)


print(f'\u2705  Helpers loaded  |  XSIAM auth: {XSIAM_AUTH_TYPE}  |  endpoint: https://{XSIAM_FQDN}')
````

### Step 4 — Test Connections

Verify all four SIEM connections before running detections.

````python
# ── XSIAM Auth Diagnostics ────────────────────────────────────────────────────
# Tries both Standard and Advanced auth and reports which one works.
# Run this cell whenever you get a 401 to identify the correct key type.
#
# Auth header reference (Cortex XDR/XSIAM API):
#   Standard : x-xdr-token = <api_key>   (no hashing)
#   Advanced : Authorization = SHA-256(api_key + nonce + timestamp_ms)  [raw hex, no prefix]
#
import hashlib, random, string, time, requests

XSIAM_FQDN_CLEAN = XSIAM_FQDN.removeprefix('https://').removeprefix('http://').rstrip('/')
url = f'https://{XSIAM_FQDN_CLEAN}/public_api/v1/xql/start_xql_query'
payload = {'request_data': {
    'query':     'dataset = xdr_data | limit 1 | fields agent_hostname',
    'timeframe': {'relativeTime': 3600000},
}}

print(f'Target URL : {url}')
print(f'Key ID     : {XSIAM_API_KEY_ID}')
print(f'Key value  : {"*" * 8}{XSIAM_API_KEY[-4:] if len(XSIAM_API_KEY) > 4 else "****"}')
print()

results = {}

# ── Test 1: Standard ─────────────────────────────────────────────────────────
print('Testing Standard auth (x-xdr-token) ...')
headers_std = {
    'x-xdr-auth-id': str(XSIAM_API_KEY_ID),
    'x-xdr-token':   XSIAM_API_KEY,
    'Content-Type':  'application/json',
}
try:
    r = requests.post(url, json=payload, headers=headers_std, timeout=15)
    results['standard'] = r.status_code
    icon = '\u2705' if r.status_code == 200 else '\u274c'
    print(f'  {icon}  HTTP {r.status_code}')
    if r.status_code != 200:
        print(f'     {r.text[:300]}')
except Exception as e:
    results['standard'] = str(e)
    print(f'  \u274c  Exception: {e}')

# ── Test 2: Advanced (raw SHA-256 digest, no prefix) ─────────────────────────
print()
print('Testing Advanced auth (SHA-256 digest, no prefix) ...')
nonce = ''.join(random.choices(string.ascii_letters + string.digits, k=64))
ts    = str(int(time.time() * 1000))
h     = hashlib.sha256((XSIAM_API_KEY + nonce + ts).encode()).hexdigest()
headers_adv = {
    'x-xdr-auth-id':   str(XSIAM_API_KEY_ID),
    'x-xdr-nonce':     nonce,
    'x-xdr-timestamp': ts,
    'Authorization':   h,       # raw SHA-256 hex — no 'Bearer' or 'Advanced' prefix
    'Content-Type':    'application/json',
}
try:
    r = requests.post(url, json=payload, headers=headers_adv, timeout=15)
    results['advanced'] = r.status_code
    icon = '\u2705' if r.status_code == 200 else '\u274c'
    print(f'  {icon}  HTTP {r.status_code}')
    if r.status_code != 200:
        print(f'     {r.text[:300]}')
except Exception as e:
    results['advanced'] = str(e)
    print(f'  \u274c  Exception: {e}')

# ── Summary ───────────────────────────────────────────────────────────────────
print()
print('=' * 55)
print('Summary')
print('=' * 55)
working = [mode for mode, code in results.items() if code == 200]
if working:
    for mode in working:
        print(f'  \u2705  {mode!r} auth works!')
        print(f'     \u2192 Set  XSIAM_AUTH_TYPE = {mode!r}  in the config cell, then re-run the helpers cell.')
else:
    print('  \u274c  Neither auth mode succeeded. Check:')
    print('       1. XSIAM_API_KEY_ID  — must be the numeric ID from the API Keys table')
    print('       2. XSIAM_API_KEY     — copied correctly with no leading/trailing spaces')
    print('       3. The key has not been revoked or expired')
    print('       4. The key has XQL / Investigation permissions assigned')
    print(f'       5. Correct tenant FQDN: {XSIAM_FQDN_CLEAN}')
````

## 🔍 Detection #1: Kernel-Mode Service Installed from Suspicious Path

**Windows Event 7045** fires when a new service (including kernel drivers) is installed via the SCM. Hunting for `type=kernel` installs outside of standard deployment paths is high-fidelity for BYOVD.

> **Note**: KDMapper bypasses the SCM — it will **not** produce Event 7045. See Detection #3.

````python
# ── XSIAM — Detection #1 ───────────────────────────────────────────────────────
QUERY = '''
dataset = xdr_data
| filter action_evtlog_event_id = 7045
| alter
    ServiceType = json_extract_scalar(action_evtlog_data_fields, "$.ServiceType"),
    ImagePath   = json_extract_scalar(action_evtlog_data_fields, "$.ImagePath"),
    ServiceName = json_extract_scalar(action_evtlog_data_fields, "$.ServiceName"),
    InstalledBy = json_extract_scalar(action_evtlog_data_fields, "$.AccountName")
| filter ServiceType ~= "kernel"
| filter ImagePath contains "Users" or ImagePath contains "Temp"
      or ImagePath contains "ProgramData" or ImagePath contains "AppData"
| fields _time, agent_hostname, ServiceName, ServiceType, ImagePath as DriverPath, InstalledBy
| sort desc _time
'''
run_xsiam(QUERY, label='Det-1 Kernel Service Install [XSIAM]')
````

````python
# ── CrowdStrike NG-SIEM — Detection #1 ─────────────────────────────────────────
QUERY = '''
#event_simpleName=DriverLoad OR #event_simpleName=ServiceInstalled
| ServiceType = /kernel/i
| ImageFileName = /(Users|Temp|ProgramData|AppData)/i
| table([timestamp, ComputerName, UserName, ServiceName, ImageFileName, SHA256HashData])
| sort(timestamp, order=desc)
'''
run_crowdstrike(QUERY, label='Det-1 Kernel Service Install [CrowdStrike]')
````

````python
# ── Elastic — Detection #1 ──────────────────────────────────────────────────────
QUERY = (
    'event.code:"7045" AND '
    'winlog.event_data.ServiceType:*kernel* AND ('
    '  winlog.event_data.ImagePath:*\\\\Users\\\\* OR '
    '  winlog.event_data.ImagePath:*\\\\Temp\\\\* OR '
    '  winlog.event_data.ImagePath:*\\\\ProgramData\\\\* OR '
    '  winlog.event_data.ImagePath:*\\\\AppData\\\\*'
    ')'
)
run_elastic(QUERY, label='Det-1 Kernel Service Install [Elastic]')
````

## 🔍 Detection #2: Known Vulnerable Driver Hash Match

Matching Sysmon Event 6 (Driver Loaded) against the [LOLDrivers](https://www.loldrivers.io/) known-vulnerable hash list is the **highest-confidence** signal available. The cell below fetches the live feed automatically.

> **Tip**: Re-run the hash refresh cell periodically to stay current with new LOLDriver disclosures.

````python
# ── Fetch live LOLDrivers hash list ────────────────────────────────────────────
import urllib.request, json as _json

LOLDRIVERS_URL = "https://www.loldrivers.io/api/drivers.json"

try:
    print(f'Fetching LOLDrivers feed from {LOLDRIVERS_URL} ...')
    with urllib.request.urlopen(LOLDRIVERS_URL, timeout=30) as resp:
        feed = _json.loads(resp.read().decode())

    VULN_HASHES = set()
    for driver in feed:
        for sample in driver.get("KnownVulnerableSamples", []):
            for field in ("SHA256", "Sha256", "sha256"):
                h = sample.get(field, "").strip().lower()
                if len(h) == 64:
                    VULN_HASHES.add(h)

    print(f'\u2705  Loaded {len(VULN_HASHES)} SHA256 hashes from LOLDrivers feed.')
except Exception as e:
    print(f'\u26a0\ufe0f  LOLDrivers feed unreachable ({e}). Using static fallback.')
    VULN_HASHES = {
        "e2e6e0b6db9428e5c8cd1a0a6f63b93c7e0ecaa4e3b2b0a0d03dedc70a24dc62",  # iqvw64.sys
        "0f16a43f7989034641fd2de3eb268bf2a10a8f0b43430220ac3aa0e547709d1f",  # zamguard64.sys
        "6c69b5c88c882e46de56e0b19e985db3a5e04a99823c8efbf0ad25d5d9f4f8d5",  # aswArPot.sys
        "b45c6136e5e3f9be6c3dbc27b2424b6d8a0b5ba7be42e2c15a4fa77d4b80caff",  # mhyprot2.sys
        "c8f9e1ad7b8ae0fbbd1f6d9be5d48a9b2a7bdae37f2c9e1e33b2f9ab84c0e6f3",  # truesight.sys
    }

print(f'   Hash list ready: {len(VULN_HASHES)} entries.')
````

````python
# ── XSIAM — Detection #2 ───────────────────────────────────────────────────────
# Build an XQL OR-chain filter from the hash set (cap at 200 to stay within query limits)
hash_filter = ' or '.join(
    f'action_module_sha256 = "{h}"' for h in sorted(VULN_HASHES)[:200]
)
QUERY = f'''
dataset = xdr_data
| filter {hash_filter}
| fields _time, agent_hostname, action_module_path,
         action_module_sha256
| sort desc _time
'''
run_xsiam(QUERY, label='Det-2 Vuln Driver Hash [XSIAM]')
````

````python
# ── CrowdStrike NG-SIEM — Detection #2 ─────────────────────────────────────────
# LogScale in-values filter
hash_vals = ", ".join(f'"{h}"' for h in sorted(VULN_HASHES)[:200])
QUERY = f'''
#event_simpleName=DriverLoad
| SHA256HashData in [{hash_vals}]
| groupBy([ComputerName, ImageFileName, SHA256HashData], function=count())
| table([_count, ComputerName, ImageFileName, SHA256HashData])
'''
run_crowdstrike(QUERY, label='Det-2 Vuln Driver Hash [CrowdStrike]')
````

````python
# ── Elastic — Detection #2 ──────────────────────────────────────────────────────
# Build a reasonable-sized Lucene OR clause (cap at 50 for query URL safety)
hash_terms = " OR ".join(f"*{h}*" for h in sorted(VULN_HASHES)[:50])
QUERY = f'event.code:"6" AND winlog.event_data.Hashes:({hash_terms})'
run_elastic(QUERY, label='Det-2 Vuln Driver Hash [Elastic]')
````

## 🔍 Detection #4: Driver Signature Enforcement (DSE) Tampering

Attackers disable DSE by modifying kernel variables (`g_CiEnabled` / `gCiOptions`) in memory, or by setting registry keys under `HKLM\SYSTEM\CurrentControlSet\Control\CI`. Some also use `bcdedit.exe /set testsigning on` to persist DSE-off across reboots.

````python
# ── XSIAM — Detection #4a: CI Registry Tamper ───────────────────────────────────
QUERY_REG = '''
dataset = xdr_data
| filter action_registry_key_name contains "Control" and action_registry_key_name contains "CI"
| fields _time, agent_hostname, actor_effective_username,
         action_registry_key_name, action_registry_value_name,
         action_registry_data, actor_process_image_path
| sort desc _time
'''
run_xsiam(QUERY_REG, label='Det-4a CI Registry [XSIAM]')

# ── XSIAM — Detection #4b: bcdedit Test-Signing ─────────────────────────────────
QUERY_BCD = '''
dataset = xdr_data
| filter action_process_command_line contains "bcdedit"
    and (action_process_command_line contains "testsigning"
      or action_process_command_line contains "bootdebug"
      or action_process_command_line contains "nointegritychecks")
| fields _time, agent_hostname, actor_effective_username,
         action_process_command_line, actor_process_image_path
| sort desc _time
'''
run_xsiam(QUERY_BCD, label='Det-4b bcdedit [XSIAM]')
````

````python
# ── CrowdStrike NG-SIEM — Detection #4 ─────────────────────────────────────────
QUERY = '''
(#event_simpleName=RegSetValue AND RegObjectName = /\\Control\\CI/i)
OR
(#event_simpleName=ProcessRollup2 AND BaseFileName = "bcdedit.exe"
  AND CommandLine = /(testsigning|bootdebug|nointegritychecks)/i)
| table([timestamp, ComputerName, UserName, RegObjectName, RegValueName,
         RegData, CommandLine, ImageFileName])
| sort(timestamp, order=desc)
'''
run_crowdstrike(QUERY, label='Det-4 DSE Tamper [CrowdStrike]')
````

````python
# ── Elastic — Detection #4 ──────────────────────────────────────────────────────
print("--- Part A: CI Registry Tamper (Sysmon Event 13) ---")
run_elastic(
    'event.code:"13" AND registry.path:*\\\\Control\\\\CI*',
    label='Det-4a CI Registry [Elastic]',
)

print("\n--- Part B: bcdedit Test-Signing ---")
run_elastic(
    'event.code:("4688" OR "1") AND process.name:bcdedit.exe AND '
    'process.args:(testsigning OR bootdebug OR nointegritychecks)',
    label='Det-4b bcdedit [Elastic]',
)
````

## 🔍 Detection #6: EDR Process Termination After Driver Load

A driver loads → within 5 minutes a known security product process is killed. This sequence is almost never legitimate.

**Monitored security processes:**
```
MsMpEng.exe, MsSense.exe, CSFalconService.exe, SentinelAgent.exe,
SentinelServiceHost.exe, xagt.exe, cb.exe, CylanceSvc.exe, ekrn.exe, avp.exe
```

````python
# ── XSIAM — Detection #6: Driver Load → EDR Kill (5-min window) ─────────────────
QUERY = '''
dataset = xdr_data
| fields _time as drv_time, agent_hostname, action_module_path as driver_path

| join type=inner (
    dataset = xdr_data
    | filter action_process_image_name ~= "(MsMpEng|MsSense|CSFalcon|SentinelAgent|SentinelService|xagt|CylanceSvc|ekrn|avp)"
        or action_process_image_name contains "cb.exe"
    | fields _time as kill_time, agent_hostname as kill_host,
             action_process_image_name as killed_proc
) as kills on agent_hostname = kill_host

| filter kill_time > drv_time and kill_time < date_add("MINUTE", 5, drv_time)
| fields drv_time, agent_hostname, driver_path, kill_time, killed_proc
| sort desc drv_time
'''
run_xsiam(QUERY, label='Det-6 Driver→EDR Kill [XSIAM]')
````

````python
# ── CrowdStrike NG-SIEM — Detection #6 ─────────────────────────────────────────
# Flag security-product process terminations that occur in bursts
QUERY = '''
#event_simpleName=ProcessRollup2
| ImageFileName = /(MsMpEng|MsSense|CSFalcon|SentinelAgent|SentinelServiceHost|xagt|CylanceSvc|ekrn|avp)/i
| groupBy([ComputerName, ImageFileName],
    function=[count(as=TermCount), min(timestamp, as=First), max(timestamp, as=Last)])
| TermCount > 1
| table([ComputerName, ImageFileName, TermCount, First, Last])
| sort(TermCount, order=desc)
'''
run_crowdstrike(QUERY, label='Det-6 EDR Kill Burst [CrowdStrike]')
````

````python
# ── Elastic — Detection #6: EQL sequence (driver load → EDR process end) ────────
es = _get_es()
eql_query = {
    "query": """
    sequence by host.name with maxspan=5m
      [any where event.code == "6"]
      [process where event.action == "end"
       and process.name : (
           "MsMpEng.exe", "MsSense.exe", "CSFalconService.exe",
           "SentinelAgent.exe", "SentinelServiceHost.exe",
           "xagt.exe", "CylanceSvc.exe", "ekrn.exe", "avp.exe"
       )
      ]
    """,
    "size": 20,
}
try:
    resp = es.eql.search(index=ES_INDEX_PATTERN, body=eql_query)
    seqs = resp.get('hits', {}).get('sequences', [])
    if seqs:
        print(f'\n\U0001f6a8  [Det-6 Elastic]  {len(seqs)} sequence(s) found!\n')
        for i, seq in enumerate(seqs[:10], 1):
            print(f'  Sequence {i}:')
            for ev in seq.get('events', []):
                src = ev.get('_source', {})
                ts  = src.get('@timestamp', '?')
                ec  = src.get('event', {}).get('code', '?')
                hn  = src.get('host', {}).get('name', '?')
                pn  = (src.get('process', {}).get('name', '') or
                       src.get('winlog', {}).get('event_data', {}).get('ImageLoaded', '?'))
                print(f'    [Event {ec}] {ts}  host={hn}  {pn}')
            print()
    else:
        print('\n\u2705  [Det-6 Elastic]  No sequences found in lookback window.')
except Exception as e:
    print(f'\u274c  EQL search failed: {e}')
````

### Hunt #5: Full BYOVD Attack Chain — .sys Write → Kernel Service → Driver Load (Elastic EQL)

````python
es = _get_es()
eql_query = {
    "query": """
    sequence by host.name with maxspan=10m
      [file where event.action in (\"creation\", \"modification\")
       and file.extension == \"sys\"
       and not file.path : (\"*\\\\System32\\\\*\", \"*\\\\SysWOW64\\\\*\",
                            \"*\\\\WinSxS\\\\*\", \"*\\\\DriverStore\\\\*\")]
      [process where event.action == \"start\"
       and process.name == \"sc.exe\"
       and process.command_line : (\"*type= kernel*\", \"*type=kernel*\")]
      [any where event.code == "6"]
    """,
    "size": 20,
}
try:
    resp = es.eql.search(index=ES_INDEX_PATTERN, body=eql_query)
    seqs = resp.get('hits', {}).get('sequences', [])
    if seqs:
        print(f'\n\U0001f6a8  [Hunt-5 BYOVD Chain Elastic]  {len(seqs)} sequence(s) found!\n')
        for i, seq in enumerate(seqs[:10], 1):
            print(f'  Sequence {i} — {len(seq.get("events", []))} events:')
            for ev in seq.get('events', []):
                src = ev.get('_source', {})
                ts  = src.get('@timestamp', '?')
                ec  = src.get('event', {}).get('code', '?')
                hn  = src.get('host', {}).get('name', '?')
                pn  = (src.get('process', {}).get('name', '') or
                       src.get('file', {}).get('path', '') or
                       src.get('winlog', {}).get('event_data', {}).get('ImageLoaded', '?'))
                print(f'    [Event {ec}] {ts}  host={hn}  {pn}')
            print()
    else:
        print('\n\u2705  [Hunt-5 BYOVD Chain Elastic]  No multi-stage sequences found.')
except Exception as e:
    print(f'\u274c  EQL search failed: {e}')
````

## 🛡️ Mitigation Recommendations

### 1. Enable HVCI (Hypervisor-Protected Code Integrity)
HVCI prevents kernel memory from being writable by non-verified code — eliminating the core primitive BYOVD exploits.
```powershell
Get-CimInstance -ClassName Win32_DeviceGuard -Namespace root\Microsoft\Windows\DeviceGuard
# HypervisorEnforcedCodeIntegrityStatus = 2  →  HVCI active
```

### 2. Deploy the Microsoft WDAC Vulnerable Driver Blocklist
```powershell
# https://learn.microsoft.com/en-us/windows/security/application-security/application-control/
#        windows-defender-application-control/design/microsoft-recommended-driver-block-rules
ConvertFrom-CIPolicy -XmlFilePath .\DriverBlock.xml -BinaryFilePath .\DriverBlock.p7b
```

### 3. Automate LOLDrivers Feed Ingestion
```bash
curl -s https://www.loldrivers.io/api/drivers.json | \\
  jq -r '.[].KnownVulnerableSamples[]?.SHA256 // empty' | sort -u > vuln_hashes.txt
```

### 4. Enable Sysmon Event 6 with Hash Enrichment
```xml
<DriverLoad onmatch="exclude">
    <Signature condition="is">Microsoft Windows</Signature>
    <Signature condition="is">Microsoft Corporation</Signature>
</DriverLoad>
```

### 5. Restrict sc.exe via WDAC / AppLocker
Alert on `sc.exe type=kernel create` from non-approved processes or paths.

### 6. Monthly Driver Fleet Audit
Inventory all running drivers with hashes and compare against LOLDrivers on a schedule.
